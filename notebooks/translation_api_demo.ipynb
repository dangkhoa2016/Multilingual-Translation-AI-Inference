## English — Run All

`DISPLAY_LANGUAGE="en"` is the default; switch to `"vi"` for Vietnamese. `DEMO_MODE="offline"` is the safe synthetic no-network mode. Select `DEMO_MODE="gpu"` only to run against **your own authenticated Modal GPU gateway**. This notebook does not deploy or start any GPU until a GPU job is deliberately submitted.

**Default benchmark:** `BENCHMARK_SEQUENCE="per_model"` warms up one model, immediately measures the same three texts on that model, then moves to the next model: **4 warm-ups + 12 measured = 16 real API jobs, 16 GPU model runs**. This avoids waiting for a different model to finish a slow cold start before measuring. `BENCHMARK_SEQUENCE="parallel"` retains the older 5-job multi-model API workflow. Both strategies use the same three inputs and can incur GPU charges.

**Cold start vs inference:** GPU-worker readiness is checked when its model server starts; the public CPU gateway `/health` does not prove that any GPU model is warm. Warm-up actually submits GPU inference and is excluded from measured results. The warm-up queue time also includes dispatch delays and startup; it does not solely measure model loading. Warm inference and queue/total time are reported separately.

**Fair and transparent comparison:** `WARMUP_TEXT` is excluded. `MEASUREMENT_TEXTS` contains three distinct public samples (short, medium and long); each model gets the same text, language pair and settings. Per-model comparison tables are presentation-only views assembled from real independent jobs, never invented API jobs. The report includes per-text timings, median, tokens/s, sampled VRAM, and **verified worker reuse** (`worker_reuse_confirmed`) when the new API provides it. Missing evidence is `N/A`, never guessed from queue times. The three texts differ in length, so medians are descriptive rather than a quality ranking.

**Privacy and credentials:** The notebook reads `TRANSLATION_GATEWAY_BASE_URL` and `TRANSLATION_API_BEARER_TOKEN` from private environment/Colab/Kaggle Secrets. When missing, Run All requests them using hidden prompts. Use only public non-sensitive text if publishing outputs. Offline results are explicitly synthetic MOCK.

**Run ID and safe reruns:** `DEMO_RUN_ID="auto"` generates a UTC timestamp and short random suffix on first execution in a Python kernel. Rerunning Run All in that **same kernel** retains the generated ID, preventing accidental duplicate paid POSTs. A **new kernel generates a new ID and can submit new paid jobs**: do not restart and replay after an uncertain POST; recover by previously accepted UUID using GET-only mode. You may deliberately set an explicit ID for a fresh evaluation. The active run ID is printed, but job UUIDs and credentials are not.

**Source text visibility:** The notebook shows the exact source input and language pair **before each request**, then displays that source input above the escaped, per-model translation table. Public notebook outputs expose the input and translations; never use confidential text in shared notebooks.

**Cost and retries:** Each real API submission has a separate deterministic idempotency key bound to `DEMO_RUN_ID` and its payload. The default per-model workflow spaces paid POST submissions to respect the default 10-job/minute limit on an otherwise idle gateway; if other clients use quota, a definite 429 still stops the demo without automatic POST replay. Repeating Run All with unchanged data in the same kernel only polls accepted jobs. After kernel restart, reuse depends on retained server idempotency records. Failed/uncertain GPU results stop later paid POSTs; GET HTTP 429 respects `Retry-After`. To read an already accepted UUID without creating a new job, use `RECOVERY_JOB_ID="prompt"` and a hidden UUID prompt.

**Warm-window recommendation:** Modal GPU workers default to a 60-second idle scaledown window (`min_containers=0`) to accommodate notebook polling, quota pacing and preflight. For longer experiments, operators can temporarily use 120 seconds, **then restore 60 seconds**. This keeps warm workers more likely to survive between rounds, but idle GPUs can incur charges. The notebook never changes Modal autoscaler settings.

```python
# Internal helpers / Các hàm nội bộ — do not edit for normal usage.
import hashlib
import json
import os
import re
import time
import uuid
from http.client import HTTPException as _HTTPClientError
from urllib.error import HTTPError
from urllib.parse import urlsplit
from urllib.request import HTTPRedirectHandler, Request, build_opener

MODEL_CONFIGS = {}
PRESERVE_FORMATTING = True
POLL_TIMEOUT_SECONDS = 1200
POLL_INTERVAL_SECONDS = 5
MAX_POLL_TRANSIENT_ERRORS = 12  # GET-only; never replay a paid POST.

def _validate_display_language():
    if DISPLAY_LANGUAGE not in ("en", "vi"):
        raise ValueError('DISPLAY_LANGUAGE must be "en" or "vi".')

def _message(english, vietnamese):
    return vietnamese if DISPLAY_LANGUAGE == "vi" else english



class APIRequestError(RuntimeError):
    def __init__(self, status_code, retry_after_seconds=None):
        self.status_code = status_code
        self.retry_after_seconds = (
            retry_after_seconds if status_code == 429 and type(retry_after_seconds) is int
            and 0 <= retry_after_seconds <= 86400 else None
        )
        message = _message(f"API returned HTTP {status_code}", f"API trả về HTTP {status_code}")
        if self.retry_after_seconds is not None:
            message += _message(f"; Retry-After: {self.retry_after_seconds} seconds.",
                                f"; Retry-After: {self.retry_after_seconds} giây; hãy đợi trước khi thử lại.")
        if status_code == 403:
            message += _message(
                "; a gateway or proxy access rule may block this client. Check Cloudflare rules or use the direct Modal HTTPS endpoint of your own deployment.",
                "; gateway hoặc proxy có thể chặn client. Kiểm tra Cloudflare hoặc dùng HTTPS endpoint Modal trực tiếp của deployment riêng.",
            )
        super().__init__(message)


class InvalidResponseError(ValueError):
    """Invalid server JSON; never echo its body."""

class NetworkRequestError(RuntimeError):
    """Transport failure, with no private URL or response body in the message."""

class NoRedirect(HTTPRedirectHandler):
    def redirect_request(self, req, fp, code, msg, headers, newurl):
        return None


_OPENER = build_opener(NoRedirect)


def _require_gateway():
    if not isinstance(API_BASE_URL, str) or not API_BASE_URL.isascii() or any(
        not 33 <= ord(char) <= 126 for char in API_BASE_URL
    ):
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng."))
    try:
        parsed = urlsplit(API_BASE_URL)
    except ValueError:
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng.")) from None
    if (
        parsed.scheme != "https"
        or not parsed.hostname
        or parsed.username
        or parsed.password
        or parsed.query
        or parsed.fragment
        or parsed.path not in ("", "/")
    ):
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng."))


def _request(method, path, *, payload=None, authenticated=False, idempotency_key=None):
    _validate_display_language()
    _require_gateway()
    if not path.startswith("/") or path.startswith("//") or "?" in path or "#" in path:
        raise ValueError(_message("Only API-relative paths are accepted.", "Chỉ chấp nhận đường dẫn tương đối của API."))
    if authenticated and not API_TOKEN:
        raise RuntimeError(_message("The private Bearer token is not configured.", "Chưa cấu hình Bearer token riêng; kiểm tra Secrets."))
    if authenticated and (not isinstance(API_TOKEN, str)
                          or not (16 <= len(API_TOKEN) <= 4096)
                          or not API_TOKEN.isascii()
                          or any(not 33 <= ord(c) <= 126 for c in API_TOKEN)):
        raise ValueError(_message(
            "Invalid private Bearer token format or length (16–4096 ASCII characters); check Secrets.",
            "Bearer token không hợp lệ về định dạng hoặc độ dài (16–4096 ký tự ASCII); kiểm tra Secrets.",
        )) from None
    if idempotency_key is not None and (
        not isinstance(idempotency_key, str) or not idempotency_key.isascii()
        or any(not 33 <= ord(c) <= 126 for c in idempotency_key)
    ):
        raise ValueError(_message("Invalid private idempotency key format; check Secrets.", "Key không hợp lệ; kiểm tra Secrets.")) from None

    # Some Cloudflare rules reject urllib's default Python User-Agent (HTTP 403).
    headers = {"Accept": "application/json",
               "User-Agent": "Multilingual-Translation-Notebook/1.0"}
    if authenticated:
        headers["Authorization"] = "Bearer " + API_TOKEN
    if idempotency_key:
        headers["Idempotency-Key"] = idempotency_key
    body = None
    if payload is not None:
        headers["Content-Type"] = "application/json"
        body = json.dumps(payload, ensure_ascii=False).encode("utf-8")
    try:
        request = Request(
            API_BASE_URL + path, data=body, headers=headers, method=method
        )
        with _OPENER.open(request, timeout=30) as response:
            try:
                data = json.load(response)
            except (ValueError, UnicodeError):
                raise InvalidResponseError(_message("Invalid JSON response; verify the API contract.", "JSON phản hồi không hợp lệ; kiểm tra tài liệu API.")) from None
            return response.status, data
    except InvalidResponseError:
        raise
    except HTTPError as exc:
        # Do not output response bodies, token, request IDs or private gateway URL.
        # A rejected manually entered token should be replaceable on the next Run All.
        if exc.code in (401, 403):
            if _NOTEBOOK_SESSION_SECRETS.get("TRANSLATION_API_BEARER_TOKEN") == API_TOKEN:
                _NOTEBOOK_SESSION_SECRETS.pop("TRANSLATION_API_BEARER_TOKEN", None)
        retry_after = None
        if exc.code == 429 and exc.headers is not None:
            raw = exc.headers.get("Retry-After", "")
            if isinstance(raw, str) and len(raw) <= 10 and raw.isascii() and raw.strip().isdigit():
                seconds = int(raw.strip())
                if 0 <= seconds <= 86400:
                    retry_after = seconds
        raise APIRequestError(exc.code, retry_after_seconds=retry_after) from None
    except (OSError, _HTTPClientError):
        raise NetworkRequestError(_message("Network request failed; check the gateway and connectivity.", "Yêu cầu mạng thất bại; kiểm tra gateway và kết nối.")) from None
    except (ValueError, UnicodeError):
        # Header/URL parsing errors can embed private values in their exception text.
        raise NetworkRequestError(_message("Invalid request encoding; check private configuration.", "Mã hóa yêu cầu không hợp lệ; kiểm tra cấu hình riêng.")) from None

# Validate job IDs and payload constraints locally.
def _job_path(job_id):
    try:
        canonical_uuid = str(uuid.UUID(str(job_id)))
    except (ValueError, AttributeError, TypeError):
        raise ValueError(_message("A valid UUID job ID is required.", "Cần UUID job hợp lệ.")) from None
    return f"/v1/jobs/{canonical_uuid}"


def _check_status_url(submission):
    expected = _job_path(submission.get("job_id"))
    if submission.get("status_url") != expected:
        raise ValueError(_message("Unexpected status_url from API.", "API trả về status_url ngoài dự kiến."))
    return expected


def _validate_payload(payload, *, multi_model):
    for field in ("source_language", "target_language"):
        value = payload.get(field)
        if not isinstance(value, str):
            raise ValueError(_message("Language must be a text code or name.", "Ngôn ngữ phải là chuỗi."))
        label = value.strip()
        if not label or len(label) > 64 or not label.isprintable() or not any(c.isalpha() for c in label):
            raise ValueError(_message("Use a printable language code or name (1–64 characters).", "Nhập mã hoặc tên ngôn ngữ in được (1–64 ký tự)."))
        payload[field] = label
    if payload["source_language"].casefold() == payload["target_language"].casefold():
        raise ValueError(_message("Source and target must differ.", "Ngôn ngữ nguồn và đích phải khác nhau."))
    text = payload.get("text")
    if not isinstance(text, str) or not text.strip() or not (1 <= len(text) <= 1000):
        raise ValueError(_message("Text must contain 1–1,000 nonblank Unicode code points.", "Văn bản phải có 1–1.000 ký tự Unicode và không được chỉ chứa khoảng trắng."))
    if "\x00" in text or any(0xD800 <= ord(char) <= 0xDFFF for char in text):
        raise ValueError(_message("Text cannot contain NUL or isolated Unicode surrogates.", "Văn bản không được chứa NUL hoặc surrogate Unicode đơn lẻ."))
    if not isinstance(payload.get("options", {}).get("preserve_formatting"), bool):
        raise ValueError(_message("preserve_formatting must be Boolean.", "preserve_formatting phải là giá trị Boolean."))
    keys = payload.get("models") if multi_model else [payload.get("model")]
    if not isinstance(keys, list) or not (1 <= len(keys) <= 4):
        raise ValueError(_message("Choose between one and four models.", "Chọn từ một đến bốn model."))
    if len(set(keys)) != len(keys) or any(key not in MODEL_KEYS for key in keys):
        raise ValueError(_message("Unknown or duplicate model key.", "Model không tồn tại hoặc bị chọn trùng."))
    configs = payload.get("model_configs", {})
    if not isinstance(configs, dict) or not set(configs) <= set(keys):
        raise ValueError(_message("Model overrides must reference selected models.",
                                  "Cấu hình riêng chỉ được áp dụng cho model đã chọn."))
    if any(not isinstance(config, dict) for config in configs.values()):
        raise ValueError(_message("Model overrides must contain option objects.",
                                  "Cấu hình riêng phải chứa các object tùy chọn."))
    return keys


def _payload(*, multi_model, text=None, models=None, model=None):
    payload = {
        "source_language": SOURCE_LANGUAGE,
        "target_language": TARGET_LANGUAGE,
        "text": TEXT_TO_TRANSLATE if text is None else text,
        "options": {"preserve_formatting": PRESERVE_FORMATTING},
    }
    if multi_model:
        payload["models"] = list(MULTI_MODEL_KEYS if models is None else models)
    else:
        payload["model"] = MODEL_KEY if model is None else model
    if MODEL_CONFIGS:
        selected = set(payload.get("models", [payload.get("model")]))
        selected_configs = {name: config for name, config in MODEL_CONFIGS.items()
                            if name in selected}
        if selected_configs:
            payload["model_configs"] = selected_configs
    _validate_payload(payload, multi_model=multi_model)
    return payload


def _request_fingerprint(path, payload, key):
    # Non-reversible binding; never output raw payload, key, gateway or token.
    canonical = json.dumps({"gateway": API_BASE_URL, "path": path, "payload": payload,
                            "key": key}, sort_keys=True, ensure_ascii=False)
    return hashlib.sha256(canonical.encode("utf-8")).hexdigest()



def _submission_preflight(selected_models):
    """Check the active gateway mode before POST; supports mock and GPU."""
    status, health = _request("GET", "/health")
    if status != 200:
        raise RuntimeError(_message("Could not check the gateway.", "Không kiểm tra được gateway."))
    phase = health.get("phase")
    if phase == "mock-only" and health.get("inference_enabled") is False:
        mode = "mock"
    elif phase == "gpu" and health.get("inference_enabled") is True:
        mode = "gpu"
    else:
        raise RuntimeError(_message(
            "Gateway mode is unavailable or inconsistent; request not sent.",
            "Chế độ gateway không hợp lệ hoặc không rõ; chưa gửi yêu cầu.",
        ))
    status, catalog = _request("GET", "/v1/models")
    if status != 200 or not isinstance(catalog, dict):
        raise RuntimeError(_message("Could not read the public model catalog.", "Không thể đọc danh mục model công khai."))
    entries = [item for item in catalog.get("models", []) if isinstance(item, dict)]
    listed = {item.get("key") for item in entries}
    if not set(selected_models).issubset(listed):
        raise RuntimeError(_message("Selected model is missing from the catalog.", "Model đã chọn không có trong danh mục."))
    if mode == "gpu":
        enabled = {item.get("key") for item in entries
                   if item.get("enabled") is True and item.get("runtime_verified") is True}
        if not set(selected_models).issubset(enabled):
            raise RuntimeError(_message(
                "Selected model is not enabled for GPU dispatch.",
                "Model chưa được bật để chạy GPU.",
            ))
        print(_message(
            "Gateway uses GPU; this request may incur inference charges.",
            "Gateway dùng GPU; yêu cầu này có thể phát sinh chi phí inference.",
        ))
    else:
        print(_message(
            "Gateway uses CPU mock; responses are simulated, not real translations.",
            "Gateway dùng CPU mock; kết quả chỉ là giả lập, không phải bản dịch thật.",
        ))
    return mode

def _validated_runs(job):
    if not isinstance(job, dict) or not isinstance(job.get("status"), str):
        raise ValueError(_message("Invalid job response structure.", "Cấu trúc phản hồi job không hợp lệ."))
    runs = job.get("runs")
    if not isinstance(runs, list) or not runs or any(
        not isinstance(run, dict) or not isinstance(run.get("model"), str)
        or not isinstance(run.get("status"), str) for run in runs
    ):
        raise ValueError(_message("Invalid job response structure.", "Cấu trúc phản hồi job không hợp lệ."))
    return runs

def _poll(job_id, *, timeout_seconds=None):
    if timeout_seconds is None:
        timeout_seconds = POLL_TIMEOUT_SECONDS  # Read the current setting when called.
    if timeout_seconds <= 0:
        raise ValueError(_message("Polling timeout must be positive.", "Thời gian chờ GET phải là số dương."))
    path = _job_path(job_id)
    started = time.monotonic()
    deadline = started + timeout_seconds
    previous = {}
    last_heartbeat = 0
    consecutive_transient_errors = 0
    terminal = {"completed", "failed", "dispatch_uncertain", "execution_uncertain"}
    safe_states = terminal | {"queued", "dispatching", "running"}
    while time.monotonic() < deadline:
        try:
            status, job = _request("GET", path, authenticated=True)
        except (APIRequestError, NetworkRequestError) as error:
            transient = (
                isinstance(error, NetworkRequestError)
                or (isinstance(error, APIRequestError)
                    and error.status_code in (429, 502, 503, 504))
            )
            if not transient:
                raise  # Authentication, conflicts, invalid payload, etc. fail fast.
            consecutive_transient_errors += 1
            if consecutive_transient_errors > MAX_POLL_TRANSIENT_ERRORS:
                raise RuntimeError(_message(
                    "Too many transient GET errors; keep the accepted job UUID and recover later.",
                    "GET lỗi tạm thời quá nhiều; giữ UUID job đã nhận và khôi phục sau.",
                )) from None
            hint = error.retry_after_seconds if isinstance(error, APIRequestError) else None
            delay = max(1, hint if hint is not None
                        else min(60, 2 ** min(consecutive_transient_errors - 1, 6)))
            remaining = deadline - time.monotonic()
            if remaining <= delay:
                raise TimeoutError(_message(
                    "Polling deadline reached during rate limiting; recover by GET-only.",
                    "Hết thời gian polling khi bị giới hạn; chỉ GET để khôi phục.",
                )) from None
            print(_message(
                f"Temporary GET failure; retry reading the same job in {delay}s.",
                f"GET bị lỗi tạm thời; đọc lại cùng job sau {delay} giây.",
            ))
            time.sleep(delay)
            continue
        if status != 200:
            raise RuntimeError(_message("Expected HTTP 200 while polling.", "GET trạng thái phải trả HTTP 200."))
        consecutive_transient_errors = 0
        try:
            returned_uuid = str(uuid.UUID(job["job_id"]))
        except (ValueError, AttributeError, TypeError, KeyError):
            raise InvalidResponseError(_message("Invalid job response identity.", "UUID trong phản hồi job không hợp lệ.")) from None
        if returned_uuid != path.rsplit("/", 1)[-1]:
            raise InvalidResponseError(_message("Job response does not match the requested UUID.", "Phản hồi job không khớp UUID đã yêu cầu.")) from None
        runs = _validated_runs(job)
        elapsed = max(0, int(time.monotonic() - started))
        changed = False
        for run in runs:
            model = run["model"] if run["model"] in MODEL_KEYS else "unknown_model"
            attempt = run.get("attempt_no", 1)
            attempt = attempt if type(attempt) is int and attempt > 0 else "unknown"
            state = run["status"] if run["status"] in safe_states else "unknown_state"
            key = (model, attempt)
            if previous.get(key) != state:
                print(_message(f"[+{elapsed}s] {model} attempt {attempt}: {state}",
                           f"[+{elapsed}s] {model} lần chạy {attempt}: {state}"))
                previous[key] = state
                changed = True
        if not changed and elapsed // 60 > last_heartbeat:
            print(_message(f"[+{elapsed}s] Waiting for job updates.", f"[+{elapsed}s] Đang chờ cập nhật."))
            last_heartbeat = elapsed // 60
        if runs and all(run.get("status") in terminal for run in runs):
            return job
        time.sleep(POLL_INTERVAL_SECONDS)
    raise TimeoutError(_message("Polling timed out. The job may still be running; do not resubmit blindly.", "GET hết thời gian, job có thể còn chạy; giữ UUID, chỉ GET lại, không tự gửi POST có phí."))



# The request implementation above is never called in offline mode.
_LIVE_REQUEST = _request
_DEMO_ACCEPTED = globals().get("_DEMO_ACCEPTED", {})
_DEMO_ATTEMPTED = globals().get("_DEMO_ATTEMPTED", set())

def _offline_request(method, path, *, payload=None, authenticated=False, idempotency_key=None):
    if method == "GET" and path == "/health":
        return 200, {"status": "ok", "inference_enabled": False, "phase": "mock-only"}
    if method == "GET" and path == "/v1/models":
        return 200, {"models": [{"key": model, "enabled": False,
                                  "runtime_verified": False, "dispatch_enabled": False}
                                 for model in MODEL_KEYS]}
    if method == "POST" and path in ("/v1/jobs/translations", "/v1/jobs/benchmarks"):
        assert payload is not None and idempotency_key
        kind = "benchmark" if path.endswith("benchmarks") else "translation"
        _validate_payload(payload, multi_model=(kind == "benchmark"))
        job_id = str(uuid.uuid5(uuid.NAMESPACE_URL, idempotency_key))
        models = payload["models"] if kind == "benchmark" else [payload["model"]]
        job = {
            "job_id": job_id, "kind": kind, "status": "completed",
            "source_language": payload["source_language"],
            "target_language": payload["target_language"],
            "runs": [{
                "run_id": str(uuid.uuid5(uuid.NAMESPACE_URL, job_id + model)),
                "model": model, "attempt_no": 1, "status": "completed",
                "translation_text": "[MOCK ONLY] Synthetic " + model +
                                     " example for " + payload["target_language"] +
                                     " — not a real translation.",
                "is_mock_result": True, "error_code": None,
            } for model in models],
        }
        _OFFLINE_JOBS[job_id] = job
        return 202, {"job_id": job_id, "kind": kind, "status": "queued",
                     "status_url": "/v1/jobs/" + job_id}
    if method == "GET" and path.startswith("/v1/jobs/"):
        job = _OFFLINE_JOBS.get(path.rsplit("/", 1)[-1])
        if job is not None:
            return 200, job
    raise RuntimeError("Unsupported offline API request (no network used)")

# Private prompts are cached only in the current Python kernel.
_NOTEBOOK_SESSION_SECRETS = globals().get("_NOTEBOOK_SESSION_SECRETS", {})


def _secret(name):
    if name not in ("TRANSLATION_GATEWAY_BASE_URL", "TRANSLATION_API_BEARER_TOKEN",
                    "TRANSLATION_EXISTING_JOB_ID"):
        raise ValueError("Unsupported private notebook setting")
    value = os.environ.get(name, "").strip()
    if value:
        return value
    try:
        from google.colab import userdata
        value = userdata.get(name)
        if isinstance(value, str) and value.strip():
            return value.strip()
    except Exception:
        pass
    try:
        from kaggle_secrets import UserSecretsClient
        value = UserSecretsClient().get_secret(name)
        if isinstance(value, str) and value.strip():
            return value.strip()
    except Exception:
        pass
    value = _NOTEBOOK_SESSION_SECRETS.get(name, "")
    if isinstance(value, str) and value.strip():
        return value.strip()
    import warnings
    from getpass import GetPassWarning, getpass
    prompt = (
        "Gateway HTTPS URL / URL HTTPS gateway" if name.endswith("BASE_URL")
        else "Accepted job UUID (hidden) / UUID job đã nhận (nhập ẩn)"
        if name == "TRANSLATION_EXISTING_JOB_ID"
        else "Bearer token (hidden) / Nhập Bearer token ẩn"
    )
    try:
        with warnings.catch_warnings():
            # Never fall back to a visible terminal input that could echo a token.
            warnings.simplefilter("error", GetPassWarning)
            value = getpass(prompt + "\n> ").strip()
    except (Exception, KeyboardInterrupt):
        raise RuntimeError(_message(
            "Private input is unavailable or cancelled. Run via a notebook UI "
            "or set the named environment variable.",
            "Không thể nhập riêng tư hoặc đã hủy. Hãy chạy từ giao diện notebook "
            "hoặc dùng biến môi trường.",
        )) from None
    if not value:
        raise RuntimeError("Required private input is empty.")
    _NOTEBOOK_SESSION_SECRETS[name] = value
    return value



def reset_demo_credentials():
    """Forget manually entered credentials before next Run All (does not POST)."""
    _NOTEBOOK_SESSION_SECRETS.clear()
    globals()["API_TOKEN"] = ""
    globals()["API_BASE_URL"] = ""
    print(_message(
        "Manual credentials cleared in this kernel; next GPU Run All will ask again.",
        "Đã xóa thông tin nhập tay khỏi kernel; GPU Run All lần sau sẽ hỏi lại.",
    ))


def accepted_demo_job_ids():
    """Read accepted UUIDs only in a trusted private session; do not publish them."""
    return tuple(sorted(item["job_id"] for item in _DEMO_ACCEPTED.values()))


def _stable_key(path, payload, *, stage=None):
    data = json.dumps({
        "run_id": DEMO_RUN_ID,
        "gateway": API_BASE_URL,
        "path": path,
        "payload": payload,
        "stage": stage,
    }, sort_keys=True, ensure_ascii=False)
    return "full-demo-" + hashlib.sha256(data.encode("utf-8")).hexdigest()

# Space paid GPU submissions instead of raising the operator's quota.
# A definite 429 still stops Run All; never automatically replay a POST.
_GPU_POST_MIN_SPACING_SECONDS = 6.25
_GPU_LAST_POST_AT = None
_gpu_monotonic = time.monotonic
_gpu_sleep = time.sleep

def _pace_gpu_submissions():
    global _GPU_LAST_POST_AT
    current = _gpu_monotonic()
    if _GPU_LAST_POST_AT is not None:
        remaining = _GPU_POST_MIN_SPACING_SECONDS - (current - _GPU_LAST_POST_AT)
        if remaining > 0:
            _gpu_sleep(remaining)
            current = _gpu_monotonic()
    _GPU_LAST_POST_AT = current


def _admit_and_poll(path, payload, *, stage=None):
    _validate_payload(payload, multi_model=path.endswith("benchmarks"))
    key = _stable_key(path, payload, stage=stage)
    fingerprint = hashlib.sha256(key.encode("ascii")).hexdigest()
    accepted = _DEMO_ACCEPTED.get(fingerprint)
    if accepted is None:
        if fingerprint in _DEMO_ATTEMPTED:
            raise RuntimeError(_message(
                "Previous POST outcome is uncertain. Recover by saved job UUID; no automatic replay.",
                "POST trước chưa rõ kết quả. Hãy đối soát UUID đã lưu; không tự gửi lại.",
            ))
        if DEMO_MODE == "gpu":
            current_mode = _submission_preflight(payload.get("models", [payload.get("model")]))
            if current_mode != "gpu":
                raise RuntimeError(_message(
                    "GPU gateway changed to CPU mock; no new job submitted.",
                    "Gateway đã chuyển sang CPU mock; không gửi job mới.",
                ))
        if DEMO_MODE == "gpu":
            _pace_gpu_submissions()
        _DEMO_ATTEMPTED.add(fingerprint)  # Before outbound POST.
        try:
            code, accepted = _request("POST", path, payload=payload,
                                      authenticated=(DEMO_MODE == "gpu"), idempotency_key=key)
        except APIRequestError as error:
            # Explicit non-admission statuses permit a deliberate, same-key retry.
            # Ambiguous failures remain guarded, never automatically re-POST.
            if error.status_code in (401, 403, 422, 429):
                _DEMO_ATTEMPTED.discard(fingerprint)
            raise
        if code != 202:
            raise RuntimeError("Expected HTTP 202 from submission")
        _check_status_url(accepted)
        _DEMO_ACCEPTED[fingerprint] = {"job_id": accepted["job_id"],
                                        "gateway": API_BASE_URL}
    elif accepted.get("gateway") != API_BASE_URL:
        raise RuntimeError("Saved submission belongs to another gateway")
    job = _poll(accepted["job_id"])
    expected_kind = "benchmark" if path.endswith("benchmarks") else "translation"
    if job.get("kind") != expected_kind:
        raise RuntimeError("Server returned a job with the wrong kind")
    return job

def _safe_run_metrics(run):
    """Numeric allowlist only; do not expose private response fields or log strings."""
    import math
    limits = {
        "usage": {"prompt_tokens": 1_000_000_000, "completion_tokens": 1_000_000_000,
                  "total_tokens": 2_000_000_000},
        "timings": {"queue_ms": 1_000_000_000, "inference_ms": 1_000_000_000,
                    "total_ms": 1_000_000_000, "prefill_ms": 1_000_000_000,
                    "decode_ms": 1_000_000_000, "tokens_per_second": 1_000_000,
                    "prompt_per_second": 1_000_000,
                    "worker_reuse_confirmed": 1, "engine_ready_ms": 1_000_000},
        "resources": {"cpu_process_seconds": 1_000_000_000,
                      "cpu_utilization_percent": 100_000,
                      "sampled_vram_mib": 200_000,
                      "sampled_gpu_utilization_percent": 100},
    }
    safe = {}
    for group, allowed in limits.items():
        values = run.get(group)
        if not isinstance(values, dict):
            continue
        for key, cap in allowed.items():
            val = values.get(key)
            if type(val) not in (int, float) or not 0 <= val <= cap:
                continue
            if not math.isfinite(val) or (group == "usage" and type(val) is not int):
                continue
            safe[key] = val
    return safe


def _latest_model_runs(job):
    """Select each model's latest positive attempt, never its first failed retry."""
    latest = {}
    for run in _validated_runs(job):
        model = run["model"]
        attempt = run.get("attempt_no", 1)
        if type(attempt) is not int or attempt < 1:
            raise InvalidResponseError(_message(
                "Invalid attempt number in API result.",
                "Số lần thử trong phản hồi API không hợp lệ.",
            ))
        previous = latest.get(model)
        if previous is None or attempt >= previous["attempt_no"]:
            latest[model] = run
    return list(latest.values())


def _require_successful_gpu_job(job, selected_models):
    """Fail closed before additional paid work if the previous job is uncertain."""
    latest = _latest_model_runs(job)
    if {run["model"] for run in latest} != set(selected_models):
        raise RuntimeError(_message(
            "GPU result is missing a requested model; do not submit another job.",
            "Kết quả GPU thiếu model đã chọn; không gửi thêm job.",
        ))
    if job.get("status") != "completed" or any(
        run["status"] != "completed" for run in latest
    ):
        raise RuntimeError(_message(
            "GPU job failed or is uncertain. Check its saved result before any new POST.",
            "GPU job thất bại hoặc chưa rõ trạng thái. Kiểm tra kết quả trước khi POST mới.",
        ))
    if any(run.get("is_mock_result") is not False for run in latest):
        raise RuntimeError(_message(
            "GPU job did not prove real inference; no additional job submitted.",
            "GPU job chưa xác nhận inference thật; không gửi job tiếp theo.",
        ))
    if any(
        not isinstance(run.get("translation_text"), str)
        or not run["translation_text"].strip()
        for run in latest
    ):
        raise RuntimeError(_message(
            "GPU response contains no translation text; stop before new paid work.",
            "Phản hồi GPU thiếu bản dịch; dừng trước khi tạo công việc có phí.",
        ))


def _comparison_html(job, *, source_text=None):
    """Accessible, escaped input and output; never render raw HTML from API."""
    from html import escape

    def safe_text(value):
        return escape(str(value), quote=True)

    # Show the original request beside the translated output, with HTML escaping.
    source_box = ""
    if isinstance(source_text, str):
        input_title = _message("English/source input", "Văn bản gốc/đầu vào")
        language_pair = f"{SOURCE_LANGUAGE} → {TARGET_LANGUAGE}"
        source_box = (
            '<div style="border:1px solid #94a3b8;border-radius:10px;'
            'padding:12px 14px;margin:12px 0;font-family:system-ui,sans-serif">'
            f'<div style="font-size:12px;font-weight:600;margin-bottom:7px">'
            f'{safe_text(input_title)} · {safe_text(language_pair)} · '
            f'{len(source_text)} chars</div>'
            '<div style="white-space:pre-wrap;overflow-wrap:anywhere;font-size:13px">'
            f'{safe_text(source_text)}</div></div>'
        )

    def cell_metric(value, unit=""):
        if value is None:
            return "N/A"
        if type(value) is int:
            return f"{value:,}" + unit
        if type(value) is float:
            return f"{value:,.2f}" + unit
        return "N/A"

    labels = (
        ("Model", "Status", "Translation", "Tokens in/out",
         "Inference", "VRAM", "GPU utilization", "Error code")
        if DISPLAY_LANGUAGE == "en"
        else ("Model", "Trạng thái", "Bản dịch", "Token vào/ra",
              "Inference", "VRAM", "Mức sử dụng GPU", "Mã lỗi")
    )
    latest = _latest_model_runs(job)
    lines = []
    for run in latest:
        metrics = _safe_run_metrics(run)
        translation = run.get("translation_text")
        output = (safe_text(translation) if isinstance(translation, str)
                  and translation else "N/A")
        status = str(run.get("status", "unknown"))
        if run.get("is_mock_result") is True:
            status += " · MOCK"
        elif (run.get("is_mock_result") is False
              and run.get("status") == "completed"
              and isinstance(run.get("translation_text"), str)
              and run["translation_text"].strip()):
            status += " · REAL"
        else:
            status += " · UNVERIFIED"
        tokens = (
            cell_metric(metrics.get("prompt_tokens")) + " / "
            + cell_metric(metrics.get("completion_tokens"))
        )
        error_code = run.get("error_code")
        safe_error = (
            safe_text(error_code)
            if isinstance(error_code, str)
            and re.fullmatch(r"[A-Za-z0-9_]{1,64}", error_code) else "N/A"
        )
        cells = (
            safe_text(run.get("model", "unknown")),
            safe_text(status), output,
            safe_text(tokens), cell_metric(metrics.get("inference_ms"), " ms"),
            cell_metric(metrics.get("sampled_vram_mib"), " MiB"),
            cell_metric(metrics.get("sampled_gpu_utilization_percent"), "%"),
            safe_error,
        )
        lines.append("<tr>" + "".join(
            f"<td>{value}</td>" for value in cells
        ) + "</tr>")
    heading = "".join(f"<th scope='col'>{safe_text(x)}</th>" for x in labels)
    return (
        source_box
        + '<div class="translation-demo-table" style="max-width:100%;overflow-x:auto;'
        'margin:14px 0;border-radius:10px;border:1px solid #cbd5e1">'
        '<table style="width:100%;border-collapse:collapse;font-size:13px;'
        'font-family:system-ui,sans-serif;text-align:left">'
        '<thead style="background:#e2e8f0;color:#0f172a">'
        f'<tr>{heading}</tr></thead>'
        '<tbody style="vertical-align:top;color:inherit">'
        + "".join(lines) + '</tbody></table></div>'
        '<style>.translation-demo-table th,.translation-demo-table td{'
        'padding:10px 12px;border-bottom:1px solid #cbd5e1;'
        'min-width:95px;overflow-wrap:anywhere;}'
        '.translation-demo-table td:nth-child(3){'
        'white-space:pre-wrap;min-width:210px;max-width:500px;}'
        '.translation-demo-table tr:last-child td{border-bottom:0;}'
        '</style>'
    )


def _show_rich_comparison(job, *, source_text=None):
    """Render in Jupyter/Colab/Kaggle; retain plain-text fallback for CLI."""
    try:
        from IPython import get_ipython
        from IPython.display import HTML, display
    except ImportError:
        return
    if get_ipython() is None:
        return
    display(HTML(_comparison_html(job, source_text=source_text)))


def _announce_request(stage, payload):
    """Show the exact public input before a possible POST, without credentials."""
    print("\n" + stage + " · " + _message("REQUEST INPUT", "VĂN BẢN YÊU CẦU"))
    print(f"  {payload['source_language']} → {payload['target_language']}")
    print("  " + payload["text"])


def _view_job(title, job, *, source_text=None):
    print("\n" + title + ": " + job["status"])
    for run in _latest_model_runs(job):
        model = run["model"] if run["model"] in MODEL_KEYS else "unknown_model"
        mock = run.get("is_mock_result")
        inference_kind = (
            "MOCK" if mock is True
            else "REAL" if (
                mock is False and run.get("status") == "completed"
                and isinstance(run.get("translation_text"), str)
                and run["translation_text"].strip()
            )
            else "UNVERIFIED"
        )
        print("  " + model + ": " + run["status"], "|", inference_kind)
        if isinstance(run.get("translation_text"), str):
            print("   ", run["translation_text"])
        error_code = run.get("error_code")
        if isinstance(error_code, str) and re.fullmatch(r"[A-Za-z0-9_]{1,64}", error_code):
            print("    error_code:", error_code)
        # Numeric allowlist only: omit private response fields and error logs.
        metrics = _safe_run_metrics(run)
        if metrics:
            print("    metrics:", metrics)
    _show_rich_comparison(job, source_text=source_text)


def _build_three_round_summary(jobs, model_keys):
    """Three observed results per selected model; report API metrics only."""
    from statistics import median

    if len(jobs) != 3:
        raise ValueError("Exactly three measured jobs are required.")
    result = {}
    fields = ("inference_ms", "queue_ms", "total_ms", "tokens_per_second",
              "prompt_tokens", "completion_tokens", "sampled_vram_mib",
              "sampled_gpu_utilization_percent", "worker_reuse_confirmed",
              "engine_ready_ms")
    for model in model_keys:
        observed = []
        for job in jobs:
            latest = {run["model"]: run for run in _latest_model_runs(job)}
            if model not in latest:
                raise ValueError("Measured job is missing a selected model.")
            run = latest[model]
            if run["status"] != "completed":
                raise RuntimeError("Cannot summarize an incomplete measured job.")
            observed.append(_safe_run_metrics(run))
        summary = {}
        for field in fields:
            numbers = [row[field] for row in observed if field in row]
            summary[field] = median(numbers) if len(numbers) == 3 else None
        result[model] = {"rounds": observed, "median": summary}
    return result


def _warmup_html(jobs):
    """Keep cold-start observations separate; queue_ms includes dispatch delays."""
    from html import escape
    labels = (
        ("Model", "Dispatch + startup", "Engine readiness", "Inference", "Total", "VRAM")
        if DISPLAY_LANGUAGE == "en"
        else ("Model", "Điều phối + khởi động", "Khởi tạo engine", "Inference", "Tổng", "VRAM")
    )
    def num(value, suffix=""):
        return f"{value:,.2f}{suffix}" if type(value) in (int, float) else "N/A"
    rows = []
    for job in jobs:
        if job is None:
            continue
        for run in _latest_model_runs(job):
            m = _safe_run_metrics(run)
            cells = (escape(str(run["model"])),
                     num(m.get("queue_ms"), " ms"),
                     num(m.get("engine_ready_ms"), " ms"),
                     num(m.get("inference_ms"), " ms"),
                     num(m.get("total_ms"), " ms"),
                     num(m.get("sampled_vram_mib"), " MiB"))
            rows.append("<tr>" + "".join(f"<td>{c}</td>" for c in cells) + "</tr>")
    th = "".join(f"<th>{escape(x)}</th>" for x in labels)
    return (
        '<div class="translation-benchmark-report" style="overflow-x:auto;'
        'border:1px solid #94a3b8;border-radius:10px;margin:12px 0">'
        '<table style="width:100%;border-collapse:collapse;font-size:13px;'
        'font-family:system-ui;color:inherit"><thead><tr>'
        f'{th}</tr></thead><tbody>{"".join(rows)}</tbody></table>'
        '</div><style>.translation-benchmark-report td,'
        '.translation-benchmark-report th{padding:9px;text-align:left;'
        'border-bottom:1px solid #94a3b855}</style>'
    )


def _view_warmup_metrics(jobs):
    label = _message(
        "Warm-up observations (excluded from measurement; queue includes startup)",
        "Quan sát warm-up (không tính vào đo; queue gồm cả khởi động)",
    )
    print("\n" + label)
    for job in jobs:
        if job is None:
            continue
        for run in _latest_model_runs(job):
            values = _safe_run_metrics(run)
            q = values.get("queue_ms")
            print(f"  {run['model']}: " + (f"queue/startup={q:.2f}ms"
                                        if q is not None else "queue/startup=N/A"))
    try:
        from IPython import get_ipython
        from IPython.display import HTML, display
        if get_ipython() is not None:
            display(HTML(_warmup_html(jobs)))
    except ImportError:
        pass


def _summary_html(summary):
    """Escaped responsive report; every round shares the same input across models."""
    from html import escape
    def format_value(value, suffix=""):
        if type(value) in (int, float):
            return f"{value:,.2f}{suffix}"
        return "N/A"

    if DISPLAY_LANGUAGE == "en":
        headers = ("Model", "Inference median", "Dispatch/start median",
                   "Total median", "Tokens/s median", "VRAM median", "Warm reuse")
        round_headers = ("Model", "Inference", "Dispatch/start", "Tokens/s", "VRAM", "Warm reuse")
        subtitle = ("Medians summarize three different text lengths, identical "
                    "between models. Queue includes dispatch and startup. "
                    "These are not controlled same-length latency measurements.")
        round_label = "Text"
    else:
        headers = ("Model", "Inference trung vị", "Điều phối/khởi động",
                   "Tổng trung vị", "Token/s trung vị", "VRAM trung vị", "Tái sử dụng warm")
        round_headers = ("Model", "Inference", "Điều phối/khởi động", "Token/s", "VRAM", "Tái sử dụng warm")
        subtitle = ("Trung vị từ ba văn bản khác độ dài, giống nhau cho mọi model. "
                    "Queue bao gồm điều phối/khởi động, không phải phép đo "
                    "latency của các văn bản cùng độ dài.")
        round_label = "Văn bản"
    head = "".join(f"<th scope='col'>{escape(h)}</th>" for h in headers)
    rows = []
    for model, data in summary.items():
        m = data["median"]
        values = (
            escape(model), format_value(m["inference_ms"], " ms"),
            format_value(m["queue_ms"], " ms"),
            format_value(m["total_ms"], " ms"),
            format_value(m["tokens_per_second"]),
            format_value(m["sampled_vram_mib"], " MiB"),
            ("N/A" if any(row.get("worker_reuse_confirmed") is None
                        for row in data["rounds"]) else
             str(sum(row.get("worker_reuse_confirmed") == 1
                     for row in data["rounds"])) + "/3"),
        )
        rows.append("<tr>" + "".join(f"<td>{v}</td>" for v in values) + "</tr>")
    body = (
        f'<p style="margin:8px 0;font-size:12px">{escape(subtitle)}</p>'
        '<div class="translation-benchmark-report" style="overflow-x:auto;'
        'border:1px solid #94a3b8;border-radius:10px;margin-bottom:12px">'
        '<table style="font-family:system-ui;font-size:13px;width:100%;'
        'border-collapse:collapse;color:inherit">'
        f'<thead><tr>{head}</tr></thead><tbody>{"".join(rows)}</tbody>'
        '</table></div>'
    )
    for idx in range(3):
        round_header = "".join(f"<th scope='col'>{escape(h)}</th>" for h in round_headers)
        per_model = []
        for model, info in summary.items():
            m = info["rounds"][idx]
            values = (
                escape(model), format_value(m.get("inference_ms"), " ms"),
                format_value(m.get("queue_ms"), " ms"),
                format_value(m.get("tokens_per_second")),
                format_value(m.get("sampled_vram_mib"), " MiB"),
                ("YES" if m.get("worker_reuse_confirmed") == 1 else
                 "NO" if m.get("worker_reuse_confirmed") == 0 else "N/A"),
            )
            per_model.append("<tr>" + "".join(f"<td>{v}</td>" for v in values) + "</tr>")
        body += (
            '<details class="translation-benchmark-report"'
            + (" open" if idx == 0 else "") + '>'
            f'<summary style="padding:8px 0;font-weight:600">'
            f'{escape(round_label)} {idx+1}/3 · {len(summary)} models</summary>'
            '<div style="overflow-x:auto;border:1px solid #94a3b8;'
            'border-radius:10px"><table style="font-family:system-ui;'
            'font-size:13px;width:100%;border-collapse:collapse;color:inherit">'
            f'<thead><tr>{round_header}</tr></thead>'
            f'<tbody>{"".join(per_model)}</tbody></table></div></details>'
        )
    return (body + '<style>.translation-benchmark-report td,'
            '.translation-benchmark-report th{padding:9px;text-align:left;'
            'border-bottom:1px solid #94a3b855}</style>')

def _view_three_round_summary(summary):
    title = _message("Three measured translations per model (median)",
                     "Ba bản dịch đo cho mỗi model (trung vị)")
    print("\n" + title)
    for model, data in summary.items():
        m = data["median"]
        elapsed = m["inference_ms"]
        print("  " + model + ": " + (f"{elapsed:.2f} ms" if elapsed is not None else "N/A"))
    try:
        from IPython import get_ipython
        from IPython.display import HTML, display
        if get_ipython() is not None:
            display(HTML(_summary_html(summary)))
    except ImportError:
        pass
```

```python
# Run All: one warm-up per selected model, then three comparable text rounds.
# Mặc định tiếng Anh; có thể đổi DISPLAY_LANGUAGE thành "vi".
_validate_display_language()
if DEMO_MODE not in ("offline", "gpu"):
    raise ValueError("DEMO_MODE must be 'offline' or 'gpu'.")
if DEMO_RUN_ID == "auto":
    _auto_run_id = globals().get("_TRANSLATION_DEMO_AUTO_RUN_ID")
    if not isinstance(_auto_run_id, str) or not re.fullmatch(
        r"demo-[0-9]{8}T[0-9]{12}Z-[0-9a-f]{6}", _auto_run_id
    ):
        from datetime import datetime, timezone
        from secrets import token_hex
        _auto_run_id = (
            "demo-" + datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
            + "-" + token_hex(3)
        )
        globals()["_TRANSLATION_DEMO_AUTO_RUN_ID"] = _auto_run_id
    DEMO_RUN_ID = _auto_run_id
if not re.fullmatch(r"[A-Za-z0-9_-]{1,64}", DEMO_RUN_ID):
    raise ValueError("DEMO_RUN_ID must contain 1–64 letters, digits, _ or -.")
if not isinstance(RECOVERY_JOB_ID, str):
    raise ValueError("RECOVERY_JOB_ID must be an empty string or a valid job UUID.")
if BENCHMARK_SEQUENCE not in ("per_model", "parallel"):
    raise ValueError("BENCHMARK_SEQUENCE must be 'per_model' or 'parallel'.")

if RECOVERY_JOB_ID:
    if DEMO_MODE != "gpu":
        raise ValueError("GET-only recovery requires DEMO_MODE='gpu'; offline never connects.")
    recovered_uuid = (
        _secret("TRANSLATION_EXISTING_JOB_ID")
        if RECOVERY_JOB_ID == "prompt" else RECOVERY_JOB_ID
    )
    recovered_path = _job_path(recovered_uuid)
    API_BASE_URL = _secret("TRANSLATION_GATEWAY_BASE_URL").rstrip("/")
    API_TOKEN = _secret("TRANSLATION_API_BEARER_TOKEN")
    _request = _LIVE_REQUEST
    _require_gateway()
    translation_result = None
    warmup_result = None
    warmup_jobs = []
    measured_api_jobs = []
    benchmark_result = None
    benchmark_results = []
    comparison_summary = None
    print(_message(
        "RECOVERY ONLY: authenticated GET; no new GPU submission.",
        "CHỈ KHÔI PHỤC: GET có xác thực, không tạo job GPU mới.",
    ))
    recovery_result = _poll(recovered_path.rsplit("/", 1)[-1])
    _view_job("GET-only recovery", recovery_result)
    print(_message("GET-ONLY RECOVERY COMPLETED", "ĐÃ KHÔI PHỤC GET-ONLY"))
else:
    recovery_result = None
    print(_message(
        f"Demo run ID: {DEMO_RUN_ID} (reuse this ID to resume the same run)",
        f"Mã lượt demo: {DEMO_RUN_ID} (giữ mã này để tiếp tục cùng lượt)",
    ))
    if (
        not isinstance(MULTI_MODEL_KEYS, list)
        or len(MULTI_MODEL_KEYS) != len(set(MULTI_MODEL_KEYS))
        or not 1 <= len(MULTI_MODEL_KEYS) <= 4
        or any(model not in MODEL_KEYS for model in MULTI_MODEL_KEYS)
        or MODEL_KEY not in MODEL_KEYS
    ):
        raise ValueError("Select 1–4 unique models and include MODEL_KEY.")
    if (
        not isinstance(MEASUREMENT_TEXTS, list)
        or len(MEASUREMENT_TEXTS) != 3
        or any(not isinstance(text, str) for text in MEASUREMENT_TEXTS)
        or len(set(MEASUREMENT_TEXTS)) != 3
    ):
        raise ValueError("Exactly three measured text inputs are required.")

    # Validate ALL payloads before the first potentially billable POST.
    primary_model = MODEL_KEY if MODEL_KEY in MULTI_MODEL_KEYS else MULTI_MODEL_KEYS[0]
    warmup_single_payload = _payload(
        multi_model=False, text=WARMUP_TEXT, model=primary_model
    )
    other_models = [model for model in MULTI_MODEL_KEYS if model != primary_model]
    warmup_others_payload = (
        _payload(multi_model=True, text=WARMUP_TEXT, models=other_models)
        if other_models else None
    )
    round_payloads = [_payload(multi_model=True, text=text) for text in MEASUREMENT_TEXTS]
    # Validate all 16 possible per-model payloads before the first paid POST.
    for selected_model in MULTI_MODEL_KEYS:
        _payload(multi_model=False, text=WARMUP_TEXT, model=selected_model)
        for measured_text in MEASUREMENT_TEXTS:
            _payload(multi_model=True, text=measured_text, models=[selected_model])
    translation_result = None
    warmup_result = None
    warmup_jobs = []
    measured_api_jobs = []
    benchmark_result = None
    benchmark_results = []
    comparison_summary = None

    if DEMO_MODE == "offline":
        API_BASE_URL = "offline://local-demo"
        API_TOKEN = ""
        _OFFLINE_JOBS = globals().get("_OFFLINE_JOBS", {})
        _request = _offline_request
        print(_message("OFFLINE: synthetic warm-up and three-round comparison, no GPU/network.",
                       "OFFLINE: warm-up giả lập và ba lượt so sánh, không gọi GPU/mạng."))
    else:
        API_BASE_URL = _secret("TRANSLATION_GATEWAY_BASE_URL").rstrip("/")
        API_TOKEN = _secret("TRANSLATION_API_BEARER_TOKEN")
        _request = _LIVE_REQUEST
        _require_gateway()
        health_status, initial_health = _request("GET", "/health")
        if (health_status != 200 or not isinstance(initial_health, dict)
                or initial_health.get("phase") != "gpu"
                or initial_health.get("inference_enabled") is not True):
            raise RuntimeError(_message(
                "Gateway is not in GPU mode. No job was submitted.",
                "Gateway chưa bật GPU. Notebook chưa gửi job.",
            ))
        print(_message(
            "GPU: one warm-up per model and three measured texts; up to 16 model runs. "
            "GPU charges may apply.",
            "GPU: mỗi model warm-up một lần rồi đo ba văn bản; tối đa 16 lượt GPU có phí.",
        ))
    health_status, health_result = _request("GET", "/health")
    catalog_status, catalog_result = _request("GET", "/v1/models")
    if (
        health_status != 200 or catalog_status != 200
        or not isinstance(health_result, dict)
        or not isinstance(catalog_result, dict)
        or not isinstance(catalog_result.get("models"), list)
    ):
        raise RuntimeError("Gateway health/catalog preflight failed")
    print("GET /health:", health_result.get("phase"))
    print("GET /v1/models:", len(catalog_result["models"]), "models")
    if DEMO_MODE == "gpu":
        enabled = {item.get("key") for item in catalog_result["models"]
                   if isinstance(item, dict) and item.get("enabled") is True
                   and item.get("runtime_verified") is True}
        if not set(MULTI_MODEL_KEYS) <= enabled:
            raise RuntimeError("Some models are not enabled/runtime_verified; no jobs submitted.")

    if BENCHMARK_SEQUENCE == "per_model":
        # Model-phase execution: no waiting for unrelated models between
        # warm-up and measurement. All inputs are identical across models.
        # Each POST remains a real, independently recoverable API job.
        runs_per_text = [[] for _ in MEASUREMENT_TEXTS]
        for selected_model in MULTI_MODEL_KEYS:
            warm_payload = _payload(
                multi_model=False, text=WARMUP_TEXT, model=selected_model
            )
            _announce_request(f"WARM-UP / {selected_model}", warm_payload)
            warm_job = _admit_and_poll(
                "/v1/jobs/translations", warm_payload,
                stage=f"warmup-{selected_model}"
            )
            warmup_jobs.append(warm_job)
            if len(warmup_jobs) == 1:
                translation_result = warm_job
            _view_job(f"WARM-UP / {selected_model}", warm_job,
                      source_text=WARMUP_TEXT)
            if DEMO_MODE == "gpu":
                _require_successful_gpu_job(warm_job, [selected_model])
            # Submit the three texts immediately after this model's warm-up.
            for round_index, measured_text in enumerate(MEASUREMENT_TEXTS):
                measured_payload = _payload(
                    multi_model=True, text=measured_text, models=[selected_model]
                )
                _announce_request(
                    f"TEXT {round_index + 1}/3 / {selected_model}", measured_payload
                )
                measured_job = _admit_and_poll(
                    "/v1/jobs/benchmarks", measured_payload,
                    stage=f"measure-{round_index + 1}-{selected_model}"
                )
                measured_api_jobs.append(measured_job)
                _view_job(
                    f"TEXT {round_index + 1}/3 / {selected_model}", measured_job,
                    source_text=measured_text
                )
                if DEMO_MODE == "gpu":
                    _require_successful_gpu_job(measured_job, [selected_model])
                runs_per_text[round_index].extend(_latest_model_runs(measured_job))

        # Presentation-only comparison views do not invent API UUIDs.
        # Real jobs remain in warmup_jobs and measured_api_jobs.
        warmup_result = (
            {"kind": "comparison_view", "status": "completed",
             "runs": [r for job in warmup_jobs[1:]
                      for r in _latest_model_runs(job)],
             "assembled_from_independent_jobs": True}
            if len(warmup_jobs) > 1 else None
        )
        benchmark_results = [
            {"kind": "comparison_view", "status": "completed",
             "runs": round_runs, "assembled_from_independent_jobs": True}
            for round_runs in runs_per_text
        ]
        benchmark_result = benchmark_results[-1]
        _view_warmup_metrics(warmup_jobs)
        for round_index, view in enumerate(benchmark_results, 1):
            _view_job(f"COMPARISON VIEW · TEXT {round_index}/3", view,
                      source_text=MEASUREMENT_TEXTS[round_index - 1])
    else:
        # Warm-up has distinct stage keys; its samples are NOT measured.
        _announce_request("WARM-UP 1 / " + primary_model, warmup_single_payload)
        translation_result = _admit_and_poll(
            "/v1/jobs/translations", warmup_single_payload, stage="warmup-single"
        )
        _view_job("WARM-UP 1 / " + primary_model, translation_result,
                  source_text=WARMUP_TEXT)
        if DEMO_MODE == "gpu":
            _require_successful_gpu_job(translation_result, [primary_model])
        if other_models:
            _announce_request("WARM-UP remaining models", warmup_others_payload)
            warmup_result = _admit_and_poll(
                "/v1/jobs/benchmarks", warmup_others_payload, stage="warmup-other-models"
            )
            _view_job("WARM-UP remaining models", warmup_result,
                      source_text=WARMUP_TEXT)
            if DEMO_MODE == "gpu":
                _require_successful_gpu_job(warmup_result, other_models)

        _view_warmup_metrics([translation_result, warmup_result])

        # All selected models see exactly the same inputs in each measured round.
        # Each stage has its own deterministic idempotency key, even for repeated text.
        for index, payload in enumerate(round_payloads, start=1):
            _announce_request(f"MEASURED ROUND {index}/3", payload)
            benchmark_result = _admit_and_poll(
                "/v1/jobs/benchmarks", payload, stage=f"measure-{index}"
            )
            benchmark_results.append(benchmark_result)
            _view_job(f"MEASURED ROUND {index}/3", benchmark_result,
                      source_text=MEASUREMENT_TEXTS[index - 1])
            if DEMO_MODE == "gpu":
                _require_successful_gpu_job(benchmark_result, MULTI_MODEL_KEYS)

    comparison_summary = _build_three_round_summary(benchmark_results, MULTI_MODEL_KEYS)
    _view_three_round_summary(comparison_summary)
    print(_message("FULL DEMO COMPLETED", "ĐÃ HOÀN THÀNH FULL DEMO"))
```

**English:** In default `per_model` mode, `warmup_jobs` contains four real translation API jobs and `measured_api_jobs` contains twelve real single-model benchmark API jobs. `translation_result` is the first warm-up. `benchmark_results` contains three **presentation-only comparisons assembled from those real jobs** by text; `benchmark_result` references the final comparison. These comparisons deliberately have no invented API UUID. In optional `parallel` mode, `benchmark_results` contains three actual multi-model API jobs. `comparison_summary` summarizes API-observed metrics; missing telemetry is `N/A`.

The notebook renders one warm-up queue/startup table and a separate warm inference comparison with expandable tables for the three input texts. Queue/startup includes dispatch delays; it is not isolated model loading time.

**GET-only recovery:** With `DEMO_MODE="gpu"` and `RECOVERY_JOB_ID="prompt"`, Run All prompts privately for an accepted job UUID and only performs a GET, even when the GPU dispatcher is off. `TRANSLATION_EXISTING_JOB_ID` can also come from private Secrets. `accepted_demo_job_ids()` returns accepted UUIDs in the current trusted kernel. Run `reset_demo_credentials()` in a temporary private code cell to replace manually entered credentials. The default auto run ID is retained within one kernel; starting a new kernel creates a new potentially billable run. Supply a new explicit `DEMO_RUN_ID` only when deliberately starting a new evaluation. Never publish UUIDs, credentials or sensitive outputs.
