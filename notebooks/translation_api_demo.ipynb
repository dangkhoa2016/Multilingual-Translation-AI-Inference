````python
def _validate_display_language():
    if globals().get("DISPLAY_LANGUAGE", "en") not in ("en", "vi"):
        raise ValueError('DISPLAY_LANGUAGE must be "en" or "vi".')


def _message(english, vietnamese):
    """Select a message safely, including while sanitizing another exception."""
    return vietnamese if globals().get("DISPLAY_LANGUAGE", "en") == "vi" else english


def _validate_exclusive_paid_post_flags():
    """Reject conflicting paid POST modes before any network operation."""
    for toggle in ("ACKNOWLEDGE_PAID_GPU", "RUN_SINGLE_MODEL_POST", "RUN_MULTI_MODEL_POST"):
        if type(globals().get(toggle, False)) is not bool:
            raise ValueError(_message(
                "Paid-run switches must be Python booleans (True or False); no request sent.",
                "Các cờ chạy có phí phải là boolean Python (True hoặc False); chưa gửi yêu cầu.",
            ))
    if bool(globals().get("RUN_SINGLE_MODEL_POST")) and bool(globals().get("RUN_MULTI_MODEL_POST")):
        raise ValueError(_message(
            "Both paid POST flags are enabled. Choose single translation OR benchmark; no request sent.",
            "Đang bật cả hai cờ POST có phí. Chỉ chọn dịch một model HOẶC benchmark; chưa gửi yêu cầu.",
        ))


# Request-key helper: first offline Run All initializes this automatically.
# This cell stays collapsed; it never sends HTTP requests.
def _configure_request_keys(kinds=("single", "benchmark")):
    global _PREVIOUS_REQUEST_KEYS, _LAST_KEY_ACTIONS, _ACTION_KEYS, _REQUEST_IDENTITIES, _SUBMISSION_HISTORY, _PAID_POST_ATTEMPTS
    # Safe advanced defaults make the short primary workflow independent of optional cells.
    defaults = {
        "MULTI_MODEL_KEYS": list(MODEL_KEYS)[:2], "RUN_MULTI_MODEL_POST": False,
        "RUN_HEALTH_GET": False, "RUN_MODEL_CATALOG_GET": False,
        "RUN_EXISTING_JOB_POLL": False, "LOAD_EXISTING_JOB_ID_FROM_SECRETS": False,
        "LOAD_PERSISTED_IDEMPOTENCY_KEYS": False, "SHOW_JOB_IDS": False,
        "SAVED_SUBMISSION_NUMBER": 0, "POLL_TIMEOUT_SECONDS": 900,
        "POLL_INTERVAL_SECONDS": 5, "NEW_BENCHMARK_REQUEST_ACTION": "",
    }
    for name, value in defaults.items():
        globals().setdefault(name, value)
    # Basic configures both request kinds; Advanced revalidates benchmark state.
    if any(kind not in ("single", "benchmark") for kind in kinds):
        raise ValueError(_message("Unsupported request kind.", "Loại yêu cầu không được hỗ trợ."))
    actions = tuple((
        kind,
        globals().get("NEW_SINGLE_REQUEST_ACTION" if kind == "single"
                      else "NEW_BENCHMARK_REQUEST_ACTION", ""),
        "SINGLE_MODEL_IDEMPOTENCY_KEY" if kind == "single"
        else "MULTI_MODEL_IDEMPOTENCY_KEY",
    ) for kind in kinds)
    if globals().get("LOAD_PERSISTED_IDEMPOTENCY_KEYS", False) and any(
        globals().get(label, "") for label in
        ("NEW_SINGLE_REQUEST_ACTION", "NEW_BENCHMARK_REQUEST_ACTION")
    ):
        raise ValueError(_message("Choose either new request keys or persisted-key recovery, not both.", "Không kết hợp nhãn yêu cầu mới với khôi phục key đã lưu."))
    _PREVIOUS_REQUEST_KEYS = globals().get("_PREVIOUS_REQUEST_KEYS", {"single": [], "benchmark": []})
    _LAST_KEY_ACTIONS = globals().get("_LAST_KEY_ACTIONS", {})
    # Stable label-to-key identity prevents accidental duplicate paid submissions.
    _ACTION_KEYS = globals().get("_ACTION_KEYS", {"single": {}, "benchmark": {}})
    for _, label, _ in actions:
        if not isinstance(label, str) or len(label) > 40 or any(
            ch not in "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789_-" for ch in label
        ):
            raise ValueError(_message("Use a short non-secret alphanumeric action label.", "Nhãn yêu cầu phải ngắn, không chứa bí mật, chỉ gồm chữ, số, _ hoặc -."))
    for kind, label, name in actions:
        if label:
            known = _ACTION_KEYS[kind].get(label)
            if known is None:
                known = globals().get(name) if _LAST_KEY_ACTIONS.get(kind) == label else None
                if known is None:
                    known = "notebook-" + ("multi" if kind == "benchmark" else "single") + "-" + uuid.uuid4().hex
                    print(_message("New " + kind + " request key created (value hidden); previous key kept in memory.",
                                   "Đã tạo key mới cho " + ("so sánh" if kind == "benchmark" else "dịch một model")
                                   + " (giá trị được ẩn); key trước vẫn được giữ trong bộ nhớ."))
                _ACTION_KEYS[kind][label] = known
            previous = globals().get(name)
            if previous and previous != known and previous not in _PREVIOUS_REQUEST_KEYS[kind]:
                _PREVIOUS_REQUEST_KEYS[kind].append(previous)
            globals()[name] = known
            _LAST_KEY_ACTIONS[kind] = label
    for kind, name in (("single", "SINGLE_MODEL_IDEMPOTENCY_KEY"),
                       ("multi", "MULTI_MODEL_IDEMPOTENCY_KEY")):
        if name not in globals():
            globals()[name] = "notebook-" + kind + "-" + uuid.uuid4().hex
    _REQUEST_IDENTITIES = globals().get("_REQUEST_IDENTITIES", {})
    # Accepted job UUIDs survive subsequent action-label changes in this kernel.
    # Fingerprints are opaque SHA-256 bindings; no text, token or key is printed.
    _SUBMISSION_HISTORY = globals().get("_SUBMISSION_HISTORY", {"single": {}, "multi": {}})
    # Same-kernel paid POST guard, including uncertain network outcomes.
    _PAID_POST_ATTEMPTS = globals().get("_PAID_POST_ATTEMPTS", set())

# Run All resets optional live paths; Basic-only edits preserve Advanced settings.
RUN_MULTI_MODEL_POST = False
RUN_HEALTH_GET = False
RUN_MODEL_CATALOG_GET = False
RUN_EXISTING_JOB_POLL = False
LOAD_EXISTING_JOB_ID_FROM_SECRETS = False
LOAD_PERSISTED_IDEMPOTENCY_KEYS = False
SHOW_JOB_IDS = False
NEW_BENCHMARK_REQUEST_ACTION = ""
````

## 02. Safety and publication / An toàn thực thi và công bố

**English**

- **Paid POST replay:** Accepted or uncertain POSTs stay blocked in the current kernel. Only confirmed HTTP 429 admission rejection permits a later *manual* retry with the original key and payload. After restart, use GET and reconcile: the in-memory guard is not durable.
- **Offline default:** All POST/GET calls start disabled; paid work requires explicit opt-in. Basic settings and both POST cells reject simultaneous paid POST flags before HTTP.
- **Private data:** Store credentials in Secrets; translated text and UUIDs are hidden by default.
- **Publishing:** Clear outputs, remove private input from source cells, and inspect copies/checkpoints. Clearing outputs **does not erase** private source text.

Read [request identity and persistent keys](#141-request-identity-and-persistent-keys--định-danh-và-lưu-khóa-yêu-cầu) before preserving keys across restarts.

**Tiếng Việt**

- **Chặn POST có phí lặp lại:** Kernel chặn POST đã nhận hoặc chưa rõ kết quả. Chỉ khi HTTP 429 xác nhận chưa nhận job, bạn mới được *tự thử lại* với key và payload gốc. Sau restart, hãy GET và đối soát; cơ chế chặn trong RAM không bền vững.
- **Mặc định ngoại tuyến:** Mọi cờ POST/GET đều tắt; cần chủ động bật để gửi job có phí. Basic settings và cả hai cell POST chặn khi bật đồng thời hai cờ có phí trước khi gọi HTTP.
- **Dữ liệu riêng tư:** Lưu credential trong Secrets; mặc định ẩn bản dịch và UUID.
- **Xuất bản:** Xóa output, loại bỏ đầu vào riêng tư khỏi source và kiểm tra bản sao/checkpoint. Xóa output **không xóa** dữ liệu trong mã nguồn.

Đọc [hướng dẫn lưu key để khôi phục](#141-request-identity-and-persistent-keys--định-danh-và-lưu-khóa-yêu-cầu) trước khi lưu key qua nhiều phiên.

### Quick reference — English

| Mode | Required configuration | Run only |
|---|---|---|
| Offline preview | **Every live HTTP flag `False`**, including all optional GET flags and both paid POST flags | Run All |
| One model | Acknowledge paid GPU; enable single POST and submitted-job polling; keep multi POST off | Basic settings → Secrets → POST translation → GET job |
| Compare models | Acknowledge paid GPU; enable multi POST and submitted-job polling; keep single POST off | Basic settings → Secrets → POST benchmark → GET job |
| Read an accepted job | Both paid POST flags `False`; enable matching GET polling and retain the accepted job reference, gateway and token | Basic settings → GET job (or GET-only recovery after restart) |

**Safety:** Optional GET flags can contact your gateway even when both paid POST flags are off. Paid POST flags may create billable jobs. Never use Run All for live requests; run only the required cells in order.

## 04. Load private Secrets / Nạp Secrets riêng tư

**English**

Run the next code cell immediately after Basic settings. It reads your selected Colab/Kaggle Secrets or inherited local environment variables **without displaying them**. It does not read `.env`. A failed reload blocks API calls until credentials are loaded successfully; accepted job references remain in memory.

**Tiếng Việt**

Chạy cell tiếp theo ngay sau Basic settings. Cell đọc Secrets đã chọn của Colab/Kaggle hoặc biến môi trường local **mà không in giá trị**. Cell không đọc `.env`. Nếu nạp lại thất bại, lời gọi API sẽ bị chặn đến khi nạp thông tin xác thực thành công; tham chiếu job đã nhận vẫn ở trong RAM.


````python
# Invalidate active credentials before any reload step can fail.
# Accepted job references and request identities remain available for recovery.
API_BASE_URL = ""
API_TOKEN = ""
EXISTING_JOB_ID = ""
_LOADED_PERSISTED_KEYS = set()

# A new request action is incompatible with persisted-key recovery.
if LOAD_PERSISTED_IDEMPOTENCY_KEYS and (
    NEW_SINGLE_REQUEST_ACTION or NEW_BENCHMARK_REQUEST_ACTION
):
    raise ValueError(_message("Choose either new request keys or persisted-key recovery, not both.", "Không dùng nhãn yêu cầu mới cùng lúc với khôi phục key đã lưu."))

# Stage provider values; failed reads must not mix old/new environment credentials.
_provider_values = {}
# Optional: load *named* private secrets from one notebook platform.
# Set the platform selection in Basic Settings; defaults are False.
if LOAD_FROM_COLAB_SECRETS and LOAD_FROM_KAGGLE_SECRETS:
    raise ValueError(_message("Select only one secrets provider.", "Chỉ chọn một nguồn Secrets."))
if LOAD_FROM_COLAB_SECRETS:
    from google.colab import userdata
    for name in ("TRANSLATION_GATEWAY_BASE_URL", "TRANSLATION_API_BEARER_TOKEN"):
        _provider_values[name] = userdata.get(name)
if LOAD_FROM_KAGGLE_SECRETS:
    from kaggle_secrets import UserSecretsClient
    client = UserSecretsClient()
    for name in ("TRANSLATION_GATEWAY_BASE_URL", "TRANSLATION_API_BEARER_TOKEN"):
        _provider_values[name] = client.get_secret(name)

# Optional third secret for recovery of a previously submitted job.
if LOAD_EXISTING_JOB_ID_FROM_SECRETS:
    if LOAD_FROM_COLAB_SECRETS:
        _provider_values["TRANSLATION_EXISTING_JOB_ID"] = userdata.get("TRANSLATION_EXISTING_JOB_ID")
    elif LOAD_FROM_KAGGLE_SECRETS:
        _provider_values["TRANSLATION_EXISTING_JOB_ID"] = client.get_secret("TRANSLATION_EXISTING_JOB_ID")
    else:
        raise ValueError(_message("Select a secrets provider to load an existing job ID.", "Chọn nguồn Secrets trước khi nạp UUID job cũ."))

if any(not isinstance(value, str) or "\0" in value for value in _provider_values.values()):
    raise ValueError(_message("Invalid private Secrets value; check the selected provider.", "Secrets không hợp lệ; kiểm tra nguồn đã chọn."))

# Restore only the key(s) for paid POST type(s) explicitly selected in Basic settings.
# GET-only recovery by an accepted UUID does not require an idempotency key.
if LOAD_PERSISTED_IDEMPOTENCY_KEYS:
    selected = []
    if RUN_SINGLE_MODEL_POST:
        selected.append(("single", "TRANSLATION_SINGLE_IDEMPOTENCY_KEY",
                         "SINGLE_MODEL_IDEMPOTENCY_KEY"))
    if RUN_MULTI_MODEL_POST:
        selected.append(("multi", "TRANSLATION_MULTI_IDEMPOTENCY_KEY",
                         "MULTI_MODEL_IDEMPOTENCY_KEY"))
    if not selected:
        print(_message("Persisted keys skipped: GET-only UUID recovery needs no idempotency key.", "Bỏ qua key đã lưu: khôi phục UUID chỉ bằng GET không cần idempotency key."))

    restored = []
    for kind, secret_name, variable_name in selected:
        if LOAD_FROM_COLAB_SECRETS:
            raw = userdata.get(secret_name)
        elif LOAD_FROM_KAGGLE_SECRETS:
            raw = client.get_secret(secret_name)
        else:
            raw = os.environ.get(secret_name)
        if not isinstance(raw, str) or not (1 <= len(raw.strip()) <= 200):
            raise ValueError(_message("Missing or invalid persisted key (1–200 chars) for a selected POST type.", "Key đã lưu bị thiếu/sai cho loại POST được chọn; kiểm tra Secrets."))
        restored.append((kind, variable_name, raw.strip()))
    if len(restored) == 2 and restored[0][2] == restored[1][2]:
        raise ValueError(_message("Supply two distinct persistent idempotency keys for both POST types.", "Cần hai key khác nhau khi gửi cả single và benchmark."))
    for kind, variable_name, value in restored:
        globals()[variable_name] = value
        _LOADED_PERSISTED_KEYS.add(kind)

# Keep explicit names visible to notebook lint/static checks after dynamic key restoration.
SINGLE_MODEL_IDEMPOTENCY_KEY = globals()["SINGLE_MODEL_IDEMPOTENCY_KEY"]
MULTI_MODEL_IDEMPOTENCY_KEY = globals()["MULTI_MODEL_IDEMPOTENCY_KEY"]

# Publish provider values only after every selected Secret has been read successfully.
os.environ.update(_provider_values)
# Set these through notebook/environment secrets; never put private values here.
API_BASE_URL = os.environ.get("TRANSLATION_GATEWAY_BASE_URL", "").strip().rstrip("/")
raw_token = os.environ.get("TRANSLATION_API_BEARER_TOKEN", "")
API_TOKEN = raw_token
EXISTING_JOB_ID = os.environ.get("TRANSLATION_EXISTING_JOB_ID", "").strip()

````

## 05. Submit a translation / Gửi yêu cầu dịch

**English**

Review the input, model and GPU budget. Set `ACKNOWLEDGE_PAID_GPU=True` and `RUN_SINGLE_MODEL_POST=True`. This cell performs preflight GETs, then submits a potentially billable job (HTTP 202). An accepted or uncertain POST stays blocked from replay within this kernel.

**Idempotency is not permanent:** Server-side protection lasts only while the database record exists. After operator purge, the same key/payload may create another paid job. Failed or disabled POSTs preserve earlier accepted references.

**Before the first paid POST:** Generated keys and accepted UUIDs exist in kernel memory only, unless you configure `LOAD_PERSISTED_IDEMPOTENCY_KEYS=True` **before submitting**. Follow [persistent-key guidance](#141-request-identity-and-persistent-keys--định-danh-và-lưu-khóa-yêu-cầu).

**HTTP 429:** An explicit admission rejection means no job was accepted. Wait for quota, then manually retry with the original key/payload and unchanged action label.

**Tiếng Việt**

Kiểm tra đầu vào, model và ngân sách GPU. Bật `ACKNOWLEDGE_PAID_GPU=True` và `RUN_SINGLE_MODEL_POST=True`. Cell thực hiện GET kiểm tra trước, sau đó gửi một job có thể phát sinh phí (HTTP 202). POST đã nhận hoặc chưa rõ kết quả sẽ bị chặn gửi lại trong kernel hiện tại.

**Idempotency không vĩnh viễn:** Server chỉ chống trùng khi bản ghi còn trong database. Sau khi operator xóa bản ghi, cùng key/payload có thể tạo job có phí khác. POST lỗi hoặc bị tắt vẫn giữ tham chiếu job đã nhận trước đó.

**Trước POST có phí đầu tiên:** Key tự sinh và UUID chỉ nằm trong RAM nếu chưa cấu hình `LOAD_PERSISTED_IDEMPOTENCY_KEYS=True` **trước khi gửi**. Xem [hướng dẫn lưu key](#141-request-identity-and-persistent-keys--định-danh-và-lưu-khóa-yêu-cầu).

**HTTP 429:** Gateway xác nhận chưa nhận job. Đợi quota khả dụng, rồi chủ động chạy lại với key/payload và nhãn yêu cầu gốc.


````python
def run_translation_cell():
    global submitted_single, single_payload
    _validate_exclusive_paid_post_flags()
    # POST translation
    # Gửi bản dịch
    submitted_single = globals().get("submitted_single")
    if RUN_SINGLE_MODEL_POST:
        if LOAD_PERSISTED_IDEMPOTENCY_KEYS and "single" not in globals().get("_LOADED_PERSISTED_KEYS", set()):
            raise RuntimeError(_message("Load the selected persisted key via Secrets before paid POST.", "Nạp key đã lưu qua Secrets trước khi POST có phí."))
        single_payload = _payload(multi_model=False)
        _require_gateway()  # Reject invalid local configuration without binding a key.
        _verify_request_identity(
            "/v1/jobs/translations", single_payload, SINGLE_MODEL_IDEMPOTENCY_KEY
        )
        if _paid_post_already_attempted("/v1/jobs/translations", single_payload, SINGLE_MODEL_IDEMPOTENCY_KEY):
            print(_message("SKIPPED repeated paid POST in this kernel; GET the saved job or reconcile an uncertain request.", "Không POST lại; chỉ GET hoặc đối soát job."))
        else:
            _gpu_preflight([single_payload["model"]])
            submitted_single = _submit(
                "/v1/jobs/translations", single_payload,
                idempotency_key=SINGLE_MODEL_IDEMPOTENCY_KEY,
            )
            submitted_single = _remember_submission("single", {
                **submitted_single, "_gateway_origin": API_BASE_URL,
                "_request_fingerprint": _request_fingerprint(
                    "/v1/jobs/translations", single_payload, SINGLE_MODEL_IDEMPOTENCY_KEY),
            })
            if SHOW_JOB_IDS:
                print(_message("PRIVATE JOB ID (clear outputs after saving):", "UUID RIÊNG TƯ (xóa output sau khi lưu):"), submitted_single["job_id"])
            else:
                print(_message("POST /v1/jobs/translations accepted (202); job ID kept in memory only.", "POST /v1/jobs/translations đã nhận (202); UUID chỉ nằm trong RAM."))
                print(_message("Before switching requests or restarting: save the UUID privately below; no extra POST.", "Trước khi đổi yêu cầu hoặc restart: lưu kín UUID bên dưới; không POST thêm."))
    else:
        print(_message("SKIPPED POST /v1/jobs/translations — explicit paid-GPU opt-in required", "BỎ QUA POST /v1/jobs/translations — cần chủ động cho phép GPU có phí."))


run_translation_cell()
````

#### Request identity and persistent keys

Every logical request is bound to its **gateway, endpoint, payload and idempotency key**. Rerunning Basic/Advanced settings in the same kernel retains existing keys. **Prepare a persistent key before a new paid POST:**

## 16. Responses and limitations / Phản hồi và giới hạn

**English**

| HTTP status | Meaning and action |
|---|---|
| `200` / `202` | GET succeeded / POST accepted. |
| `401` | Verify the private Bearer token and access permissions. |
| `404` | Check the saved UUID and original gateway. |
| `409` | The idempotency key was reused with a different payload. |
| `422` | Validate model, language labels and Unicode input length locally. |
| `429` | Respect a valid numeric `Retry-After` (0–86,400 seconds). For a **POST admission rejection**, wait for quota and manually retry with the same key/payload; never retry automatically. For a **GET**, wait, then GET again. |
| `dispatch_uncertain` / `execution_uncertain` | Reconcile the job; never automatically retry POST. |

Polling timeout does not prove the job failed. Keep the UUID and issue GET later. HTTP error bodies are intentionally hidden. An uncertain POST requires operator reconciliation; after database purge even the original key could create a new paid job.

This notebook **does not deploy GPU infrastructure or score translation quality**. The public template contains no saved execution outputs. See [recovery](#14-advanced-usage-and-recovery--sử-dụng-nâng-cao-và-khôi-phục).

**Tiếng Việt**

| Trạng thái HTTP | Ý nghĩa và cách xử lý |
|---|---|
| `200` / `202` | GET thành công / POST đã được nhận. |
| `401` | Kiểm tra Bearer token riêng tư và quyền truy cập. |
| `404` | Kiểm tra UUID đã lưu và gateway gốc. |
| `409` | Idempotency key đã được dùng với payload khác. |
| `422` | Kiểm tra model, nhãn ngôn ngữ và độ dài Unicode của đầu vào. |
| `429` | Tuân thủ `Retry-After` dạng số hợp lệ (0–86.400 giây). Nếu **POST bị từ chối ở bước tiếp nhận**, đợi quota và chỉ chủ động thử lại với key/payload cũ; không tự retry. Với **GET**, đợi rồi GET lại. |
| `dispatch_uncertain` / `execution_uncertain` | Đối soát job; không tự động thử lại POST. |

GET timeout không chứng minh job thất bại. Hãy giữ UUID rồi GET lại sau. Body lỗi HTTP được ẩn để bảo vệ dữ liệu. POST chưa rõ kết quả phải được operator đối soát; sau khi xóa database record, key gốc vẫn có thể tạo job có phí mới.

Notebook **không triển khai GPU hoặc chấm chất lượng dịch**. Template công khai không chứa output đã chạy. Xem [khôi phục](#14-advanced-usage-and-recovery--sử-dụng-nâng-cao-và-khôi-phục).

### Client helpers / Các hàm hỗ trợ client

**Các hàm hỗ trợ client**

**English**

The next four code cells were initialized by the initial offline **Run All**. They are collapsed by default in JupyterLab. **Do not edit or rerun them for normal API usage.** Return to [POST translation](#05-submit-a-translation--gửi-yêu-cầu-dịch) or [GET job](#07-read-translation-results--đọc-kết-quả-dịch). If you intentionally modify the client implementation, initialize all four helper cells again **in order**.

**Tiếng Việt**

Bốn cell code tiếp theo đã được khởi tạo trong lần **Run All ngoại tuyến** đầu tiên và mặc định được thu gọn trong JupyterLab. **Không sửa hoặc chạy lại khi dùng API thông thường.** Hãy quay về [POST translation](#05-submit-a-translation--gửi-yêu-cầu-dịch) hoặc [GET job](#07-read-translation-results--đọc-kết-quả-dịch). Nếu chủ động thay đổi mã client, phải khởi tạo lại cả bốn cell helper **theo đúng thứ tự**.


````python
from http.client import HTTPException as _HTTPClientError


class APIRequestError(RuntimeError):
    def __init__(self, status_code, retry_after_seconds=None):
        self.status_code = status_code
        self.retry_after_seconds = (
            retry_after_seconds if status_code == 429 and type(retry_after_seconds) is int
            and 0 <= retry_after_seconds <= 86400 else None
        )
        message = _message(f"API returned HTTP {status_code}", f"API trả về HTTP {status_code}")
        if self.retry_after_seconds is not None:
            message += _message(f"; Retry-After: {self.retry_after_seconds} seconds.",
                                f"; Retry-After: {self.retry_after_seconds} giây; hãy đợi trước khi thử lại.")
        super().__init__(message)


class InvalidResponseError(ValueError):
    """Invalid server JSON; never echo its body."""

class NetworkRequestError(RuntimeError):
    """Transport failure, with no private URL or response body in the message."""

class NoRedirect(HTTPRedirectHandler):
    def redirect_request(self, req, fp, code, msg, headers, newurl):
        return None


_OPENER = build_opener(NoRedirect)


def _require_gateway():
    if not isinstance(API_BASE_URL, str) or not API_BASE_URL.isascii() or any(
        not 33 <= ord(char) <= 126 for char in API_BASE_URL
    ):
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng."))
    try:
        parsed = urlsplit(API_BASE_URL)
    except ValueError:
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng.")) from None
    if (
        parsed.scheme != "https"
        or not parsed.hostname
        or parsed.username
        or parsed.password
        or parsed.query
        or parsed.fragment
        or parsed.path not in ("", "/")
    ):
        raise ValueError(_message("Set a valid HTTPS gateway root URL via private environment variables.", "Hãy cấu hình HTTPS gateway hợp lệ qua Secrets riêng."))


def _request(method, path, *, payload=None, authenticated=False, idempotency_key=None):
    _validate_display_language()
    _require_gateway()
    if not path.startswith("/") or path.startswith("//") or "?" in path or "#" in path:
        raise ValueError(_message("Only API-relative paths are accepted.", "Chỉ chấp nhận đường dẫn tương đối của API."))
    if authenticated and not API_TOKEN:
        raise RuntimeError(_message("The private Bearer token is not configured.", "Chưa cấu hình Bearer token riêng; kiểm tra Secrets."))
    if authenticated and (not isinstance(API_TOKEN, str)
                          or not (16 <= len(API_TOKEN) <= 4096)
                          or not API_TOKEN.isascii()
                          or any(not 33 <= ord(c) <= 126 for c in API_TOKEN)):
        raise ValueError(_message(
            "Invalid private Bearer token format or length (16–4096 ASCII characters); check Secrets.",
            "Bearer token không hợp lệ về định dạng hoặc độ dài (16–4096 ký tự ASCII); kiểm tra Secrets.",
        )) from None
    if idempotency_key is not None and (
        not isinstance(idempotency_key, str) or not idempotency_key.isascii()
        or any(not 33 <= ord(c) <= 126 for c in idempotency_key)
    ):
        raise ValueError(_message("Invalid private idempotency key format; check Secrets.", "Key không hợp lệ; kiểm tra Secrets.")) from None

    headers = {"Accept": "application/json"}
    if authenticated:
        headers["Authorization"] = "Bearer " + API_TOKEN
    if idempotency_key:
        headers["Idempotency-Key"] = idempotency_key
    body = None
    if payload is not None:
        headers["Content-Type"] = "application/json"
        body = json.dumps(payload, ensure_ascii=False).encode("utf-8")
    try:
        request = Request(
            API_BASE_URL + path, data=body, headers=headers, method=method
        )
        with _OPENER.open(request, timeout=30) as response:
            try:
                data = json.load(response)
            except (ValueError, UnicodeError):
                raise InvalidResponseError(_message("Invalid JSON response; verify the API contract.", "JSON phản hồi không hợp lệ; kiểm tra tài liệu API.")) from None
            return response.status, data
    except InvalidResponseError:
        raise
    except HTTPError as exc:
        # Do not output response bodies, token, request IDs or private gateway URL.
        retry_after = None
        if exc.code == 429 and exc.headers is not None:
            raw = exc.headers.get("Retry-After", "")
            if isinstance(raw, str) and len(raw) <= 10 and raw.isascii() and raw.strip().isdigit():
                seconds = int(raw.strip())
                if 0 <= seconds <= 86400:
                    retry_after = seconds
        raise APIRequestError(exc.code, retry_after_seconds=retry_after) from None
    except (OSError, _HTTPClientError):
        raise NetworkRequestError(_message("Network request failed; check the gateway and connectivity.", "Yêu cầu mạng thất bại; kiểm tra gateway và kết nối.")) from None
    except (ValueError, UnicodeError):
        # Header/URL parsing errors can embed private values in their exception text.
        raise NetworkRequestError(_message("Invalid request encoding; check private configuration.", "Mã hóa yêu cầu không hợp lệ; kiểm tra cấu hình riêng.")) from None
````

````python
# Validate job IDs and payload constraints locally.
def _job_path(job_id):
    try:
        canonical_uuid = str(uuid.UUID(str(job_id)))
    except (ValueError, AttributeError, TypeError):
        raise ValueError(_message("A valid UUID job ID is required.", "Cần UUID job hợp lệ.")) from None
    return f"/v1/jobs/{canonical_uuid}"


def _check_status_url(submission):
    expected = _job_path(submission.get("job_id"))
    if submission.get("status_url") != expected:
        raise ValueError(_message("Unexpected status_url from API.", "API trả về status_url ngoài dự kiến."))
    return expected


def _validate_payload(payload, *, multi_model):
    for field in ("source_language", "target_language"):
        value = payload.get(field)
        if not isinstance(value, str):
            raise ValueError(_message("Language must be a text code or name.", "Ngôn ngữ phải là chuỗi."))
        label = value.strip()
        if not label or len(label) > 64 or not label.isprintable() or not any(c.isalpha() for c in label):
            raise ValueError(_message("Use a printable language code or name (1–64 characters).", "Nhập mã hoặc tên ngôn ngữ in được (1–64 ký tự)."))
        payload[field] = label
    if payload["source_language"].casefold() == payload["target_language"].casefold():
        raise ValueError(_message("Source and target must differ.", "Ngôn ngữ nguồn và đích phải khác nhau."))
    text = payload.get("text")
    if not isinstance(text, str) or not text.strip() or not (1 <= len(text) <= 1000):
        raise ValueError(_message("Text must contain 1–1,000 nonblank Unicode code points.", "Văn bản phải có 1–1.000 ký tự Unicode và không được chỉ chứa khoảng trắng."))
    if "\x00" in text or any(0xD800 <= ord(char) <= 0xDFFF for char in text):
        raise ValueError(_message("Text cannot contain NUL or isolated Unicode surrogates.", "Văn bản không được chứa NUL hoặc surrogate Unicode đơn lẻ."))
    if not isinstance(payload.get("options", {}).get("preserve_formatting"), bool):
        raise ValueError(_message("preserve_formatting must be Boolean.", "preserve_formatting phải là giá trị Boolean."))
    keys = payload.get("models") if multi_model else [payload.get("model")]
    if not isinstance(keys, list) or not (1 <= len(keys) <= 4):
        raise ValueError(_message("Choose between one and four models.", "Chọn từ một đến bốn model."))
    if len(set(keys)) != len(keys) or any(key not in MODEL_KEYS for key in keys):
        raise ValueError(_message("Unknown or duplicate model key.", "Model không tồn tại hoặc bị chọn trùng."))
    return keys


def _payload(*, multi_model):
    payload = {
        "source_language": SOURCE_LANGUAGE,
        "target_language": TARGET_LANGUAGE,
        "text": TEXT_TO_TRANSLATE,
        "options": {"preserve_formatting": PRESERVE_FORMATTING},
    }
    if multi_model:
        payload["models"] = list(MULTI_MODEL_KEYS)
    else:
        payload["model"] = MODEL_KEY
    _validate_payload(payload, multi_model=multi_model)
    return payload


def _request_fingerprint(path, payload, key):
    # Non-reversible binding; never output raw payload, key, gateway or token.
    canonical = json.dumps({"gateway": API_BASE_URL, "path": path, "payload": payload,
                            "key": key}, sort_keys=True, ensure_ascii=False)
    return hashlib.sha256(canonical.encode("utf-8")).hexdigest()


````

### GPU dispatch and submission / Điều phối GPU và gửi yêu cầu

**Điều phối GPU và gửi yêu cầu**

**English**

Preflight GETs inspect gateway mode and catalog metadata (`enabled` and `runtime_verified`). They do **not** prove live GPU worker health, warm state or readiness. Cold starts and GPU failures remain possible. Billable POSTs perform preflight GETs even when the optional GET display flags are disabled.

**Tiếng Việt**

GET kiểm tra trước chỉ đọc chế độ gateway và metadata danh mục model (`enabled` và `runtime_verified`). Các thông tin này **không** xác nhận GPU worker đang khỏe, đã nạp model hoặc sẵn sàng. Vẫn có thể gặp cold start hoặc lỗi GPU. POST có phí tự thực hiện GET kiểm tra dù cờ hiển thị GET tùy chọn đang tắt.


````python
def _gpu_preflight(selected_models):
    if not ACKNOWLEDGE_PAID_GPU:
        raise RuntimeError(_message("Paid requests require ACKNOWLEDGE_PAID_GPU and a POST flag.", "Yêu cầu có phí cần ACKNOWLEDGE_PAID_GPU và cờ POST."))
    status, health = _request("GET", "/health")
    if status != 200 or health.get("inference_enabled") is not True or health.get("phase") != "gpu":
        raise RuntimeError(_message("Gateway does not confirm configured GPU dispatch (not a worker health check).", "Gateway chưa xác nhận chế độ GPU dispatch (không phải kiểm tra worker đang khỏe)."))
    status, catalog = _request("GET", "/v1/models", authenticated=True)
    if status != 200:
        raise RuntimeError(_message("Could not verify the authenticated model catalog.", "Không thể kiểm tra danh mục model đã xác thực."))
    enabled = {m["key"] for m in catalog.get("models", [])
               if m.get("enabled") is True and m.get("runtime_verified") is True}
    if not set(selected_models).issubset(enabled):
        raise RuntimeError(_message("Selected model is not enabled/runtime_verified in catalog metadata (not a live GPU health check).", "Model chưa bật hoặc chưa xác thực runtime trong catalog (không phải kiểm tra GPU trực tiếp)."))


def _verify_request_identity(path, payload, idempotency_key):
    # Read-only check: do not bind unused keys before gateway validation or preflight.
    if (not isinstance(idempotency_key, str) or not (1 <= len(idempotency_key) <= 200)
            or not idempotency_key.isascii()
            or any(not 33 <= ord(char) <= 126 for char in idempotency_key)):
        raise ValueError(_message("An idempotency key of 1–200 characters is required.", "Cần idempotency key dài 1–200 ký tự."))
    identity = json.dumps({"gateway": API_BASE_URL, "path": path, "payload": payload},
                          sort_keys=True, ensure_ascii=False)
    prior = _REQUEST_IDENTITIES.get(idempotency_key)
    if prior is not None and prior != identity:
        raise ValueError(_message("Payload changed under the same key (or gateway/endpoint differs). For a NEW job choose a label in Basic settings (single or comparison); For uncertain POST keep the original payload, gateway and key.", "Job mới chọn nhãn trong Basic (single hoặc comparison); nếu POST chưa rõ kết quả, giữ nguyên payload, gateway và key."))
    return identity


def _check_request_identity(path, payload, idempotency_key):
    # _submit records intent after successful preflight, immediately before POST.
    identity = _verify_request_identity(path, payload, idempotency_key)
    _REQUEST_IDENTITIES[idempotency_key] = identity


def _paid_post_already_attempted(path, payload, idempotency_key):
    return _request_fingerprint(path, payload, idempotency_key) in globals().get("_PAID_POST_ATTEMPTS", set())


def _submit(path, payload, *, idempotency_key):
    _validate_exclusive_paid_post_flags()
    if _paid_post_already_attempted(path, payload, idempotency_key):
        raise RuntimeError(_message("Repeated paid POST blocked within this kernel; use GET-only recovery or reconcile an uncertain request.", "Đã chặn POST có phí lặp lại; hãy khôi phục bằng GET hoặc đối soát yêu cầu chưa rõ kết quả."))
    _check_request_identity(path, payload, idempotency_key)
    # Record BEFORE network I/O: an exception does not prove the server rejected the job.
    globals().setdefault("_PAID_POST_ATTEMPTS", set()).add(
        _request_fingerprint(path, payload, idempotency_key)
    )
    try:
        status, submitted = _request(
            "POST", path, payload=payload, authenticated=True, idempotency_key=idempotency_key
        )
    except APIRequestError as error:
        if error.status_code == 429:
            # This gateway rejects quota admission before creating/dispatching a job.
            # Release only this confirmed rejection; every ambiguous outcome stays guarded.
            _PAID_POST_ATTEMPTS.discard(_request_fingerprint(path, payload, idempotency_key))
            print(_message("POST rejected (429): no job accepted. Wait for quota availability, then manually rerun this POST with the SAME gateway, key and payload; no automatic retry.", "Quota từ chối, chưa tạo job. Chờ quota cho phép rồi tự chạy lại cell POST với CÙNG gateway, key và dữ liệu; không tự retry."))
        raise
    if status != 202:
        raise RuntimeError(_message("Expected HTTP 202 from job submission.", "Gửi job phải trả HTTP 202."))
    _check_status_url(submitted)
    return submitted

def _remember_submission(kind, accepted):
    global SAVED_SUBMISSION_NUMBER, RUN_EXISTING_JOB_POLL
    # Store only a validated UUID and opaque provenance, never raw text/key/token.
    _SUBMISSION_HISTORY.setdefault(kind, {})[accepted["_request_fingerprint"]] = {
        "job_id": accepted["job_id"],
        "_gateway_origin": accepted["_gateway_origin"],
        "_request_fingerprint": accepted["_request_fingerprint"],
    }
    # An accepted POST selects the current request; failed POSTs preserve recovery.
    SAVED_SUBMISSION_NUMBER = 0
    RUN_EXISTING_JOB_POLL = False
    return accepted


def _matching_submission(kind):
    # Explicit read-only recovery by accepted-submission order, independent of
    # the current action label/key. Numbering starts at 1 for each request kind.
    number = globals().get("SAVED_SUBMISSION_NUMBER", 0)
    if type(number) is not int or number < 0:
        raise ValueError(_message("SAVED_SUBMISSION_NUMBER must be 0 or a positive integer.", "SAVED_SUBMISSION_NUMBER phải là 0 hoặc số nguyên dương."))
    if number:
        # A historical reference is only selectable in GET-only mode; a
        # changed but unsubmitted paid request must not silently poll old work.
        if globals().get("RUN_SINGLE_MODEL_POST") or globals().get("RUN_MULTI_MODEL_POST"):
            return None, "request_mismatch"
        remembered = list(globals().get("_SUBMISSION_HISTORY", {}).get(kind, {}).values())
        if number > len(remembered):
            return None, "request_mismatch"
        candidate = remembered[number - 1]
        if not isinstance(candidate, dict) or candidate.get("_gateway_origin") != API_BASE_URL:
            return None, "gateway_mismatch"
        try:
            _job_path(candidate.get("job_id"))
        except (ValueError, TypeError, AttributeError):
            return None, "request_mismatch"
        return candidate, None
    path, key, multi = (
        ("/v1/jobs/translations", SINGLE_MODEL_IDEMPOTENCY_KEY, False)
        if kind == "single" else
        ("/v1/jobs/benchmarks", MULTI_MODEL_IDEMPOTENCY_KEY, True)
    )
    try:
        fingerprint = _request_fingerprint(path, _payload(multi_model=multi), key)
    except (ValueError, TypeError, KeyError):
        fingerprint = None
    # Prefer the original accepted submission when switching back to a label.
    history = globals().get("_SUBMISSION_HISTORY", {}).get(kind, {})
    candidate = history.get(fingerprint) if fingerprint is not None else None
    if candidate is None:
        candidate = globals().get("submitted_" + kind)
    if candidate is None:
        return None, None
    if not isinstance(candidate, dict) or candidate.get("_gateway_origin") != API_BASE_URL:
        return None, "gateway_mismatch"
    if not fingerprint or candidate.get("_request_fingerprint") != fingerprint:
        return None, "request_mismatch"
    return candidate, None


def _select_submitted_jobs(*, only_kind=None):
    if SUBMITTED_JOB_TO_POLL not in ("single", "multi", "both"):
        raise ValueError(_message("Choose single, multi or both for submitted-job polling.", "Chọn single, multi hoặc both để đọc trạng thái job."))
    kinds = (("single", "multi") if SUBMITTED_JOB_TO_POLL == "both"
             else (SUBMITTED_JOB_TO_POLL,))
    if only_kind is not None:
        if only_kind not in ("single", "multi"):
            raise ValueError(_message("Invalid job kind for GET-only selection.", "Loại job không hợp lệ cho khôi phục chỉ bằng GET."))
        kinds = (only_kind,) if only_kind in kinds else ()
    selected, errors = [], []
    for kind in kinds:
        submission, failure = _matching_submission(kind)
        if failure == "gateway_mismatch":
            errors.append({"reason": failure})
            print(_message("GET job skipped: saved submission gateway differs or is unknown; "
                           "reference preserved. Restore its gateway before polling.",
                           "Bỏ qua GET: gateway khác hoặc không xác định; vẫn giữ tham chiếu job. "
                           "Khôi phục gateway gốc trước khi polling."))
        elif failure == "request_mismatch":
            errors.append({"reason": failure})
            print(_message("GET job skipped: no accepted submission matches the current request; reference preserved for explicit UUID recovery. For uncertain POST, do not rotate the key or automatically resubmit.", "Không có job phù hợp; nếu POST chưa rõ kết quả thì không đổi key hay gửi lại."))
        elif submission is not None:
            selected.append(submission["job_id"])
    return selected, errors

````

````python
def _validated_runs(job):
    if not isinstance(job, dict) or not isinstance(job.get("status"), str):
        raise ValueError(_message("Invalid job response structure.", "Cấu trúc phản hồi job không hợp lệ."))
    runs = job.get("runs")
    if not isinstance(runs, list) or not runs or any(
        not isinstance(run, dict) or not isinstance(run.get("model"), str)
        or not isinstance(run.get("status"), str) for run in runs
    ):
        raise ValueError(_message("Invalid job response structure.", "Cấu trúc phản hồi job không hợp lệ."))
    return runs

def _poll(job_id, *, timeout_seconds=None):
    if timeout_seconds is None:
        timeout_seconds = POLL_TIMEOUT_SECONDS  # Read the current setting when called.
    if timeout_seconds <= 0:
        raise ValueError(_message("Polling timeout must be positive.", "Thời gian chờ GET phải là số dương."))
    path = _job_path(job_id)
    started = time.monotonic()
    deadline = started + timeout_seconds
    previous = {}
    last_heartbeat = 0
    terminal = {"completed", "failed", "dispatch_uncertain", "execution_uncertain"}
    safe_states = terminal | {"queued", "dispatching", "running"}
    while time.monotonic() < deadline:
        status, job = _request("GET", path, authenticated=True)
        if status != 200:
            raise RuntimeError(_message("Expected HTTP 200 while polling.", "GET trạng thái phải trả HTTP 200."))
        try:
            returned_uuid = str(uuid.UUID(job["job_id"]))
        except (ValueError, AttributeError, TypeError, KeyError):
            raise InvalidResponseError(_message("Invalid job response identity.", "UUID trong phản hồi job không hợp lệ.")) from None
        if returned_uuid != path.rsplit("/", 1)[-1]:
            raise InvalidResponseError(_message("Job response does not match the requested UUID.", "Phản hồi job không khớp UUID đã yêu cầu.")) from None
        runs = _validated_runs(job)
        elapsed = max(0, int(time.monotonic() - started))
        changed = False
        for run in runs:
            model = run["model"] if run["model"] in MODEL_KEYS else "unknown_model"
            attempt = run.get("attempt_no", 1)
            attempt = attempt if type(attempt) is int and attempt > 0 else "unknown"
            state = run["status"] if run["status"] in safe_states else "unknown_state"
            key = (model, attempt)
            if previous.get(key) != state:
                print(_message(f"[+{elapsed}s] {model} attempt {attempt}: {state}",
                           f"[+{elapsed}s] {model} lần chạy {attempt}: {state}"))
                previous[key] = state
                changed = True
        if not changed and elapsed // 60 > last_heartbeat:
            print(_message(f"[+{elapsed}s] Waiting for job updates.", f"[+{elapsed}s] Đang chờ cập nhật."))
            last_heartbeat = elapsed // 60
        if runs and all(run.get("status") in terminal for run in runs):
            return job
        time.sleep(POLL_INTERVAL_SECONDS)
    raise TimeoutError(_message("Polling timed out. The job may still be running; do not resubmit blindly.", "GET hết thời gian, job có thể còn chạy; giữ UUID, chỉ GET lại, không tự gửi POST có phí."))


def _safe_run_metrics(run):
    """Numeric allowlist only; do not expose private response fields or log strings."""
    import math
    limits = {
        "usage": {"prompt_tokens": 1_000_000_000, "completion_tokens": 1_000_000_000,
                  "total_tokens": 2_000_000_000},
        "timings": {"queue_ms": 1_000_000_000, "inference_ms": 1_000_000_000,
                    "total_ms": 1_000_000_000, "prefill_ms": 1_000_000_000,
                    "decode_ms": 1_000_000_000, "tokens_per_second": 1_000_000,
                    "prompt_per_second": 1_000_000},
        "resources": {"cpu_process_seconds": 1_000_000_000,
                      "cpu_utilization_percent": 100_000,
                      "sampled_vram_mib": 200_000,
                      "sampled_gpu_utilization_percent": 100},
    }
    safe = {}
    for group, allowed in limits.items():
        values = run.get(group)
        if not isinstance(values, dict):
            continue
        for key, cap in allowed.items():
            val = values.get(key)
            if type(val) not in (int, float) or not 0 <= val <= cap:
                continue
            if not math.isfinite(val) or (group == "usage" and type(val) is not int):
                continue
            safe[key] = val
    return safe


def _metric_label(value):
    if value is None:
        return "N/A"
    if type(value) is int:
        return f"{value:,}"
    return f"{value:,.2f}".rstrip("0").rstrip(".")


def _summary_data(job):
    runs = _validated_runs(job)
    safe_states = {"queued", "dispatching", "running", "completed", "failed",
                   "dispatch_uncertain", "execution_uncertain"}
    known_errors = {"DISPATCH_UNKNOWN", "DISPATCH_STALE_UNVERIFIED", "EXECUTION_STALE_UNVERIFIED",
                    "GPU_MODEL_MISMATCH", "GPU_INFERENCE_FAILED", "GPU_EXECUTION_UNCERTAIN",
                     "UNSUPPORTED_LANGUAGE_PAIR"}
    latest = {}
    for run in runs:
        if type(run.get("attempt_no", 1)) is not int or run.get("attempt_no", 1) < 1:
            raise ValueError(_message("Invalid attempt number in job response.", "Số lần chạy trong phản hồi không hợp lệ."))
        if run.get("translation_text") is not None and not isinstance(run["translation_text"], str):
            raise ValueError(_message("Invalid translation text in job response.", "Văn bản dịch trong phản hồi không hợp lệ."))
        latest[run["model"]] = max(latest.get(run["model"], 0), run.get("attempt_no", 1))
    summary = {
        "kind": job.get("kind") if job.get("kind") in ("translation", "benchmark") else "unknown_kind",
        "status": job["status"] if job["status"] in safe_states | {"partial_failed"} else "unknown_state",
        "runs": [
            {
                "model": run["model"] if run["model"] in MODEL_KEYS else "unknown_model",
                "attempt_no": run.get("attempt_no", 1),
                "is_latest_attempt": run.get("attempt_no", 1) == latest[run["model"]],
                "status": run["status"] if run["status"] in safe_states else "unknown_state",
                "is_mock_result": run.get("is_mock_result") if type(run.get("is_mock_result")) is bool else None,
                "translation_codepoints": len(run.get("translation_text") or ""),
                "metrics": _safe_run_metrics(run),
                "error_code": (run.get("error_code") if run.get("error_code") is None
                               or (isinstance(run.get("error_code"), str) and run["error_code"] in known_errors)
                               else "unknown_error"),
            }
            for run in runs
        ],
    }
    return summary


def _comparison_columns(job):
    summary = _summary_data(job)
    latest = {}
    for row, run in zip(summary["runs"], _validated_runs(job)):
        if row["is_latest_attempt"]:
            latest[row["model"]] = (row, run.get("translation_text"))
    order = [model for model in MODEL_KEYS if model in latest]
    if "unknown_model" in latest:
        order.append("unknown_model")
    return summary, [latest[model] for model in order]


def _comparison_html(job, *, show_translations=False):
    # Text is absent from HTML entirely unless explicitly enabled, never hidden in CSS.
    from html import escape
    _, columns = _comparison_columns(job)
    labels = {"hy_mt2_30b": "Hy-MT2", "index_35b": "Index Translate",
              "gemma4_31b_it": "Gemma 4", "qwen35_35b": "Qwen3.5"}
    cell_style = "border:1px solid #d1d5db;padding:12px;vertical-align:top;text-align:left"
    width = max(540, 150 + 240 * len(columns))
    output = [f'<div style="overflow-x:auto"><table style="border-collapse:collapse;width:100%;min-width:{width}px">',
              '<caption style="text-align:left;padding:8px 0;font-weight:bold">'
              + _message("Translation comparison", "So sánh bản dịch") + '</caption>',
              f'<thead><tr><th scope="col" style="{cell_style}">Model</th>']
    for row, _ in columns:
        key = row["model"]
        output.append(f'<th scope="col" style="{cell_style}">{escape(labels.get(key, key))}<br><code>{escape(key)}</code></th>')
    output.append('</tr></thead><tbody>')
    fields = (
        (_message("Status", "Trạng thái"), lambda row: row["status"]),
        (_message("Attempt", "Lần chạy"), lambda row: str(row["attempt_no"])),
        (_message("Execution", "Thực thi"), lambda row: _message("Mock", "Giả lập") if row["is_mock_result"] is True
         else _message("Nonmock", "Không giả lập") if row["is_mock_result"] is False else _message("Unknown", "Chưa rõ")),
        (_message("Characters", "Ký tự"), lambda row: str(row["translation_codepoints"])),
        (_message("Error code", "Mã lỗi"), lambda row: row["error_code"] or "—"),
    )
    for label, value in fields:
        output.append(f'<tr><th scope="row" style="{cell_style}">{escape(label)}</th>')
        for row, _ in columns:
            output.append(f'<td style="{cell_style}">{escape(value(row))}</td>')
        output.append('</tr>')
    if globals().get("SHOW_METRICS", True) is True:
        groups = (
            (_message("Token usage", "Số token"), (
                (_message("Input tokens", "Token đầu vào"), "prompt_tokens"),
                (_message("Output tokens", "Token đầu ra"), "completion_tokens"),
                (_message("Total tokens", "Tổng token"), "total_tokens"))),
            (_message("Timing (ms except rates)", "Thời gian (ms, trừ tốc độ)"), (
                (_message("Queue (ms)", "Chờ (ms)"), "queue_ms"),
                (_message("Inference (ms)", "Suy luận (ms)"), "inference_ms"),
                (_message("Total (ms)", "Tổng (ms)"), "total_ms"),
                (_message("Prefill (ms)", "Prefill (ms)"), "prefill_ms"),
                (_message("Decode (ms)", "Decode (ms)"), "decode_ms"),
                (_message("Tokens / second", "Token / giây"), "tokens_per_second"),
                (_message("Prompt tokens / second", "Token đầu vào / giây"), "prompt_per_second"))),
            (_message("CPU process and post-inference GPU sample",
                      "CPU tiến trình và mẫu GPU sau inference"), (
                (_message("CPU process (s)", "CPU tiến trình (giây)"), "cpu_process_seconds"),
                (_message("CPU utilization (%)", "Sử dụng CPU (%)"), "cpu_utilization_percent"),
                (_message("VRAM sample (MiB)", "Mẫu VRAM (MiB)"), "sampled_vram_mib"),
                (_message("GPU utilization sample (%)", "Mẫu GPU (%)"),
                 "sampled_gpu_utilization_percent"))),
        )
        for heading, metrics in groups:
            output.append(f'<tr><th scope="colgroup" colspan="{len(columns)+1}" '
                          f'style="{cell_style};font-weight:bold">{escape(heading)}</th></tr>')
            for label, key in metrics:
                output.append(f'<tr><th scope="row" style="{cell_style}">{escape(label)}</th>')
                for row, _ in columns:
                    output.append(f'<td style="{cell_style}">'
                                  f'{escape(_metric_label(row["metrics"].get(key)))}</td>')
                output.append('</tr>')
    output.append(f'<tr><th scope="row" style="{cell_style}">{_message("Translation", "Bản dịch")}</th>')
    for _, translated in columns:
        if translated is None:
            text = _message("No translation", "Chưa có bản dịch")
        elif show_translations is not True:
            text = _message("Hidden — SHOW_TRANSLATIONS=True", "Đang ẩn — SHOW_TRANSLATIONS=True")
        else:
            text = translated or _message("Empty output", "Đầu ra rỗng")
        output.append(f'<td style="{cell_style}"><pre style="white-space:pre-wrap;overflow-wrap:anywhere;font-family:inherit;margin:0">{escape(text)}</pre></td>')
    output.append('</tr></tbody></table></div>')
    return "".join(output)


def _display_comparison_html(markup):
    try:
        from IPython import get_ipython
        from IPython.display import HTML, display
    except ImportError:
        return False
    if get_ipython() is None:
        return False
    display(HTML(markup))
    return True


def _show_summary(job):
    summary, columns = _comparison_columns(job)
    rich = _display_comparison_html(_comparison_html(job, show_translations=SHOW_TRANSLATIONS))
    if not rich:
        # JSON fallback honors the same visibility setting as the HTML table.
        # Display a separate copy; preserve internal summary and API response.
        display_summary = summary if globals().get("SHOW_METRICS", True) is True else {
            **summary,
            "runs": [{key: value for key, value in row.items() if key != "metrics"}
                     for row in summary["runs"]],
        }
        print(json.dumps(display_summary, indent=2, ensure_ascii=False))
    print(_message("Job status uses latest attempt per model.", "Trạng thái job dùng lần thử mới nhất của mỗi model."))
    if any(row["is_mock_result"] is True for row in summary["runs"]):
        print(_message("WARNING: Mock output is not a real translation.", "CẢNH BÁO: Đầu ra giả lập không phải bản dịch thật."))
    if SHOW_TRANSLATIONS is True:
        print(_message("CAUTION: Translation text below may be saved in notebook output.", "LƯU Ý: Nội dung dịch bên dưới có thể được lưu trong output notebook."))
        if not rich:
            for row, translated in columns:
                if translated is not None:
                    print(row["model"], ":", translated)


def _run_get_job(*, only_kind=None):
    """GET-only: reuse the same safe polling and error handling for both views."""
    # Select submissions explicitly. Never silently include a previous benchmark.
    jobs_to_poll = []
    poll_failures = []  # Non-sensitive per-job outcomes; no URLs, UUIDs or exception text.
    if RUN_SUBMITTED_JOB_POLL:
        selected_ids, origin_errors = _select_submitted_jobs(only_kind=only_kind)
        jobs_to_poll.extend(selected_ids)
        poll_failures.extend(origin_errors)
    if only_kind is None and RUN_EXISTING_JOB_POLL:
        if not EXISTING_JOB_ID:
            raise ValueError(_message("Set TRANSLATION_EXISTING_JOB_ID privately before polling.", "Cần UUID job được lưu kín trước khi GET."))
        jobs_to_poll.append(EXISTING_JOB_ID)

    job_results = []
    if jobs_to_poll:
        for job_id in dict.fromkeys(jobs_to_poll):
            try:
                result = _poll(job_id)
                _show_summary(result)
            except APIRequestError as error:
                # HTTP status is safe to show; never print response bodies or exception text.
                status_code = error.status_code
                status_code = status_code if type(status_code) is int and 100 <= status_code <= 599 else None
                poll_failures.append({"reason": "http", "status_code": status_code})
                if status_code in (401, 403):
                    action = _message("check your private Bearer token and access permissions", "kiểm tra Bearer token riêng và quyền truy cập")
                elif status_code == 404:
                    action = _message("verify the saved job UUID and gateway; do not guess IDs", "kiểm tra UUID đã lưu và gateway; không đoán ID")
                elif status_code == 429:
                    action = _message("respect rate limits and wait before GET polling again", "tuân thủ giới hạn tần suất và chờ trước khi GET lại")
                elif status_code is not None and status_code >= 500:
                    action = _message("check gateway availability and retry only GET after recovery", "kiểm tra gateway và chỉ thử lại GET sau khi khôi phục")
                else:
                    action = _message("check the API contract and gateway configuration", "kiểm tra API contract và cấu hình gateway")
                label = "HTTP " + str(status_code) if status_code is not None else "HTTP error"
                print(_message("GET job " + label + "; " + action + "; reference preserved; do not resubmit paid POST.",
                               "GET job " + label + ": Giữ UUID, kiểm tra quyền/gateway và chỉ GET lại khi phù hợp, không POST mới."))
                delay = error.retry_after_seconds
                if type(delay) is int and 0 <= delay <= 86400:
                    print(_message(f"Retry-After: {delay} seconds; retry only GET manually.",
                               f"Retry-After: {delay} giây; chỉ tự GET lại, không POST mới."))
                continue
            except TimeoutError:
                poll_failures.append({"reason": "timeout"})
                print(_message("GET job timeout; job may still run. Retry only GET later; do not resubmit paid POST.", "GET quá thời gian; job có thể vẫn chạy. Chỉ GET lại, không gửi POST có phí."))
                continue
            except (NetworkRequestError, URLError, OSError):
                poll_failures.append({"reason": "network"})
                print(_message("GET job network error; check connectivity and retry only GET; do not resubmit paid POST.", "Lỗi mạng; kiểm tra kết nối và chỉ thử GET lại."))
                continue
            except (ValueError, TypeError):
                poll_failures.append({"reason": "invalid_response"})
                print(_message("GET job invalid response; verify the API contract; do not resubmit paid POST.", "Response không hợp lệ; kiểm tra tài liệu API, không gửi POST mới."))
                continue
            except RuntimeError:
                poll_failures.append({"reason": "client_error"})
                print(_message("GET job client/runtime error; check local configuration; do not resubmit paid POST.", "Lỗi client/runtime khi GET; kiểm tra cấu hình local, không gửi lại POST có phí."))
                continue
            job_results.append(result)
    else:
        print(_message("SKIPPED GET /v1/jobs/{job_id} — polling flags disabled or no job selected", "BỎ QUA GET /v1/jobs/{job_id} — cờ GET đang tắt hoặc chưa chọn job."))
    return job_results, poll_failures
````

## 19. Synthetic API responses / Phản hồi API giả lập

**English**

**SYNTHETIC EXAMPLE — NOT A REAL GPU RESULT.** The next cell constructs fictional submission and completed-job responses entirely offline. Live translation text remains hidden unless `SHOW_TRANSLATIONS=True`.

**VÍ DỤ GIẢ LẬP — KHÔNG PHẢI KẾT QUẢ GPU THẬT.** Cell tiếp theo tạo response gửi job và hoàn thành hoàn toàn giả lập, không gọi API.
